```python
"""
What each variable represents. Example values are mostly from RID 22496, 2025-02-26. Variables ending in _iAll contain data for all interations with length equal to "iterations".

n_measurements = 2000: directly from dataset
iterations = 5: the length of scan_sequence1, directly from dataset
n_excitation_cycles = [0,  2, 20,  8,  2,  8, ...], length = 1 + n_measurements * iterations: the number of excitation cycles in each measurement
n_excitation_cycles_iAll = [16830, 16665, 16488, 16194, 16221], length = iterations: total number of excitation cycles in each iteration as an array.
n_excitation_attempts = 1: directly from dataset. Number of excitation pulses in each excitation cycle.
n_tStamps_iAll = [16830, 16665, 16488, 16194, 16221], length = iterations: total number of timestamps in each iteration as an array.
SPCM0_Photon_tStamps = [[0, 0], [-1, -1], [-1, -1], [1.32237714e+06, -1], [-1, -1], ...], length = sum(n_tStamps_iAll) = 82399, for example. Directly from dataset.
SPCM0_tStamps_i =  [[-1, -1], [-1, -1], [1.32237714e+06, -1], [-1, -1], ...], length = elements of n_tStamps_iAll, 16830 for the first iteration, for example. The timestamps in each iteration.
reference_tStamps_t1 = [0, 1322341.65973562, 1322341.65976682, 1322341.65979802, ...], length = sum(n_tStamps_iAll) = 82399, for example. Directly from dataset.
t1_reference_i = [1322341.65973562, 1322341.65976682, 1322341.65979802, ...], length = elements of n_tStamps_iAll, 16830 for the first iteration, for example. The reference times in each iteration.
SPCM0_1click_counts_iAll = [128., 122., 172., 186., 172.], length = iterations: number of 1-click events in each iteration as an array; events in the first column of SPCM0_tStamps_i for different iteration.
SPCM0_2click_counts_iAll = [0., 0., 0., 0., 0.], length = iterations: number of 2-click events in each iteration as an array; events in the second column of SPCM0_tStamps_i for different iteration.
start_indices = [0, 16830, 33495, 49983, 66177], length = iterations: to slice arrays like SPCM0_Photon_tStamps into iterations.
    Remember that SPCM0_Photon_tStamps is a single array with the timestamps for all iterations.
end_indices = [16830, 33495, 49983, 66177, 82398], length = iterations: same as start_indices
SPCM0_rel_tStamps_i = [1099.89196062, 1088.94892037, 1098.02931547, ...], length = SPCM0_1click_counts_iAll + SPCM0_2click_counts_iAll, for example length is 128 and 122 for first two iterations.
    This array contains the timestamps relative to the reference time t1 in ns unit. -1 event are excluded and 2-click events are included.
SPCM0_rel_tStamps_iAll = [[1099.89196062, 1088.94892037, 1098.02931547, ...], [..], ..], length = iterations. Arrays of arrays containing all the relative timestamps for all iteration.
AllSPCMs_tStamps_filtered_i = [[1757511.19383094, 1757533.11171639, ...], length = total number of photons in iteration i,
    i.e. length = all 1-click and 2-click counts from SPCM0, SPCM1, SPCM0_OtherNode, and SPCM1_OtherNode. Contains all the absolute time stamps from all 4 SPCMs with -1 removed and sorted from small to large.
AllSPCMs_tStamps_filtered_iAll = [[1757511.19383094, 1757533.11171639, ...], [1749316.29104084, 1749328.33163165, ..], ..], length = iterations. Array of arrays keeping all
    AllSPCMs_tStamps_filtered_i from all iterations.
Ave_photon_rate_iAll = [[0.11084697, 0.1147433 , 0.10952382, 0.1337287 , 0.12123424]], length = iterations. Contains the average photon generation rate for all iterations.
n_coincidence_iAll = [0,0,1,0,2], length = iterations. The number of coincidences in each iteration used to calculate g2.

Other notes:
1- Does this work if I change max_clicks from 2 to 1 or 3? The answer is no.

"""
show_photon_hist = False
### Print the title of the output
display(HTML(f"""<b style='font-size:18px;'> {date_filters[0]} - RID {rid} - {experiment_function}</b>"""))

Col_Eff = 0.05 # overal expected/theoretical collection and detection efficiency

### Instead of using a large n_measurements, which can sometimes lead to underflow error, we use dummy_variable to split the measurements.
### So, if scan_variable1_name == 'dummy_variable', we want to combine all iterations into one:
if scan_variable1_name == 'dummy_variable':
    n_measurements = n_measurements * iterations
    iterations = 1
    scan_sequence1 = [0]

## To look at specific iterations, you can select in scan_sequence_1, for example scan_sequence1 = [0,2]. For now, the first element has to be 0.
# scan_sequence1 = [0,4]
# iterations = len(scan_sequence1)
# n_measurements = n_measurements * iterations


n_excitation_cycles_iAll = np.zeros(iterations, dtype=int) ### total number of excitation cycle in each iteration.
n_tStamps_iAll = np.zeros(iterations, dtype=int)  ### number of timestamps in each iteration.
for i in range(iterations):
    n_excitation_cycles_iAll[i] = sum(n_excitation_cycles[i*n_measurements+1:(i+1)*n_measurements+1]+1)
    if scan_variable1_name == 'n_excitation_attempts':
        n_excitation_attempts = scan_sequence1[i]
    n_tStamps_iAll[i] = n_excitation_cycles_iAll[i] * n_excitation_attempts
    ### Note that n_excitation_cycles varies in each cycle, each contains n_excitation_attempts.
    ### the first +1 is because the first element in each dataset is 0.
    ### the last +1 is because there is one more cycle than n_excitation_cycles in each loop. See the oscciloscope. For example, if n_excitation_cycles[j] = 2,
    ### there are 2+1=3 sets of excitation cycles for that measurement.


SPCM0_1click_counts_iAll = np.array([])
SPCM0_2click_counts_iAll = np.array([])
SPCM1_1click_counts_iAll = np.array([])
SPCM1_2click_counts_iAll = np.array([])
SPCM0_OtherNode_1click_counts_iAll = np.array([])
SPCM0_OtherNode_2click_counts_iAll = np.array([])
SPCM1_OtherNode_1click_counts_iAll = np.array([])
SPCM1_OtherNode_2click_counts_iAll = np.array([])

AllSPCMs_tStamps_filtered_iAll = []

SPCM0_rel_tStamps_iAll = []
SPCM1_rel_tStamps_iAll = []
SPCM0_OtherNode_rel_tStamps_iAll = []
SPCM1_OtherNode_rel_tStamps_iAll = []

n_coincidence_iAll = np.zeros(iterations)
g2_0 = np.zeros(iterations)
g2_0_err = np.zeros(iterations)
Ave_photon_rate_iAll = np.zeros(iterations)


def get_rel_tStamps_i(tStamps_i, t1_reference_i, spcm_name):
    ### the timestamps relative to reference time. -1 events are excluded.
    rel_tStamps_i_temp = [row[row != -1] - ref for row, ref in zip(tStamps_i, t1_reference_i)]
    if not rel_tStamps_i_temp: ### np.concatenate crashes on empty lists; if no photon detected, I replace timetags by [0].
        rel_tStamps_i_temp = [[0]]
    rel_tStamps_i_temp2 = np.concatenate(rel_tStamps_i_temp) * 1e9 ## Flatten to combine 1-click and 2-click events.
    rel_tStamps_i = rel_tStamps_i_temp2[rel_tStamps_i_temp2 >= 0]
    ## in 1 out of 1million case, the reference time is later than the timetag and I get a negative
    ### element which messes up the histograms. I am simply removing these elements if any, and printing
    if len(rel_tStamps_i) != len(rel_tStamps_i_temp2):
        print(f'Early time stamp for {len(rel_tStamps_i_temp2) - len(rel_tStamps_i)} elements on {spcm_name} was detected.')
    return rel_tStamps_i


def get_tStamps_i(tStamps_all, start, end):
    tStamps_i = np.asarray(tStamps_all[1:][start:end])
    if tStamps_i.size == 0:
        return np.empty((0, 2))
    if tStamps_i.ndim == 1:
        return tStamps_i.reshape(-1, 2)
    return tStamps_i


### Calculate the ending and starting indices for the iterations
end_indices = np.cumsum(n_tStamps_iAll)
start_indices = np.insert(end_indices[:-1], 0, 0) ### prepend a 0 for the starting index

for i in range(iterations):
    start = start_indices[i]
    end = end_indices[i]
    SPCM0_tStamps_i = get_tStamps_i(SPCM0_Photon_tStamps, start, end)
    SPCM1_tStamps_i = get_tStamps_i(SPCM1_Photon_tStamps, start, end)
    SPCM0_OtherNode_tStamps_i = get_tStamps_i(SPCM0_OtherNode_Photon_tStamps, start, end)
    SPCM1_OtherNode_tStamps_i = get_tStamps_i(SPCM1_OtherNode_Photon_tStamps, start, end)
    t1_reference_i = reference_tStamps_t1[1:][start:end]

    ### number of 1-click events (the first column is none -1)
    SPCM0_1click_counts_iAll = np.append(SPCM0_1click_counts_iAll, np.count_nonzero(SPCM0_tStamps_i[:, 0] != -1))
    SPCM1_1click_counts_iAll = np.append(SPCM1_1click_counts_iAll, np.count_nonzero(SPCM1_tStamps_i[:, 0] != -1))
    SPCM0_OtherNode_1click_counts_iAll = np.append(SPCM0_OtherNode_1click_counts_iAll, np.count_nonzero(SPCM0_OtherNode_tStamps_i[:, 0] != -1))
    SPCM1_OtherNode_1click_counts_iAll = np.append(SPCM1_OtherNode_1click_counts_iAll, np.count_nonzero(SPCM1_OtherNode_tStamps_i[:, 0] != -1))

    ### number of 2-click events (the 2nd column is none -1)
    SPCM0_2click_counts_iAll = np.append(SPCM0_2click_counts_iAll, np.count_nonzero(SPCM0_tStamps_i[:, 1] != -1))
    SPCM1_2click_counts_iAll = np.append(SPCM1_2click_counts_iAll, np.count_nonzero(SPCM1_tStamps_i[:, 1] != -1))
    SPCM0_OtherNode_2click_counts_iAll = np.append(SPCM0_OtherNode_2click_counts_iAll, np.count_nonzero(SPCM0_OtherNode_tStamps_i[:, 1] != -1))
    SPCM1_OtherNode_2click_counts_iAll = np.append(SPCM1_OtherNode_2click_counts_iAll, np.count_nonzero(SPCM1_OtherNode_tStamps_i[:, 1] != -1))

    SPCM0_rel_tStamps_i = get_rel_tStamps_i(SPCM0_tStamps_i, t1_reference_i, 'SPCM0')
    SPCM1_rel_tStamps_i = get_rel_tStamps_i(SPCM1_tStamps_i, t1_reference_i, 'SPCM1')
    SPCM0_OtherNode_rel_tStamps_i = get_rel_tStamps_i(SPCM0_OtherNode_tStamps_i, t1_reference_i, 'SPCM0_OtherNode')
    SPCM1_OtherNode_rel_tStamps_i = get_rel_tStamps_i(SPCM1_OtherNode_tStamps_i, t1_reference_i, 'SPCM1_OtherNode')

    SPCM0_rel_tStamps_iAll.append(SPCM0_rel_tStamps_i)
    SPCM1_rel_tStamps_iAll.append(SPCM1_rel_tStamps_i)
    SPCM0_OtherNode_rel_tStamps_iAll.append(SPCM0_OtherNode_rel_tStamps_i)
    SPCM1_OtherNode_rel_tStamps_iAll.append(SPCM1_OtherNode_rel_tStamps_i)

    ### calculate g2 for each iteration. In future, we can also select a narrower coincidence window.
    ### For 4 SPCMs, count coincidences between all unordered detector pairs.
    first_click_arrays_i = [
        SPCM0_tStamps_i[:, 0],
        SPCM1_tStamps_i[:, 0],
        SPCM0_OtherNode_tStamps_i[:, 0],
        SPCM1_OtherNode_tStamps_i[:, 0],
    ]
    first_click_counts_i = [
        SPCM0_1click_counts_iAll[i],
        SPCM1_1click_counts_iAll[i],
        SPCM0_OtherNode_1click_counts_iAll[i],
        SPCM1_OtherNode_1click_counts_iAll[i],
    ]

    n_coincidence_i = 0
    g2_denominator_i = 0
    g2_denominator_var_i = 0
    for j in range(len(first_click_arrays_i)):
        for k in range(j + 1, len(first_click_arrays_i)):
            n_coincidence_i += np.sum((first_click_arrays_i[j] >= 1) & (first_click_arrays_i[k] >= 1))

            Nj = first_click_counts_i[j]
            Nk = first_click_counts_i[k]
            g2_denominator_i += Nj * Nk
            g2_denominator_var_i += (Nk**2) * Nj + (Nj**2) * Nk

    # n_coincidence_i = 1 ### if there is no coincidence, you can set n_coincidence_i = 1 manually to find the bound to g2 because the next count might be a coincidence.
    n_coincidence_iAll[i] = n_coincidence_i ### Having a list of coincidences for each iteration helps to learn more about the g2.

    if g2_denominator_i > 0:
        g2_0_temp = (n_coincidence_i / g2_denominator_i) * n_excitation_cycles_iAll[i]
        ### Errors for g2 results are from error propagation assuming Poissonian distribution for counts:
        if n_coincidence_i > 0:
            g2_0_err_temp = g2_0_temp * np.sqrt(1/n_coincidence_i + g2_denominator_var_i/(g2_denominator_i**2))
        else:
            g2_0_err_temp = np.nan
    else:
        g2_0_temp = np.nan
        g2_0_err_temp = np.nan
    g2_0[i] = g2_0_temp
    g2_0_err[i] = g2_0_err_temp

    ############## absolute timestamps of all the photons for rate analysis
    AllSPCMs_tStamps_flatten_i = np.concatenate((
        SPCM0_tStamps_i,
        SPCM1_tStamps_i,
        SPCM0_OtherNode_tStamps_i,
        SPCM1_OtherNode_tStamps_i,
    )).flatten() ### flatten to get the timestamps of all photons (1st and 2nd clicks).
    AllSPCMs_tStamps_filtered_i = AllSPCMs_tStamps_flatten_i[AllSPCMs_tStamps_flatten_i != -1]  ### -1 events are removed
    AllSPCMs_tStamps_filtered_i = np.sort(AllSPCMs_tStamps_filtered_i) ### sorted from small to large timestamps
    AllSPCMs_tStamps_filtered_iAll.append(AllSPCMs_tStamps_filtered_i)


    # ###########checking H1H2 ###############
    # count=0
    # for j in range(len(SPCM0_tStamps_i)):
    #     if SPCM0_tStamps_i[:, 0][j] != -1 and SPCM0_OtherNode_tStamps_i[:, 0][j] != -1:
    #         count+=1
    
    # print("H1H2 ", count)
    # ###########checking V1V2 ###############
    # count=0
    # for j in range(len(SPCM1_tStamps_i)):
    #     if SPCM1_tStamps_i[:, 0][j] != -1 and SPCM1_OtherNode_tStamps_i[:, 0][j] != -1:
    #         count+=1
    
    # print("V1V2 ",count)
    
    # ###########checking (H1orV1) (H2orV2) ###############
    # count=0
    # for j in range(len(SPCM1_tStamps_i)):
    #     if (SPCM0_tStamps_i[:, 0][j] != -1 or SPCM1_tStamps_i[:, 0][j] != -1) and (SPCM0_OtherNode_tStamps_i[:, 0][j] != -1 or SPCM1_OtherNode_tStamps_i[:, 0][j] != -1):
    #         count+=1
    
    # print("(H1orV1) (H2orV2) ",count)
    


AllSPCMs_1click_counts_iAll = (
    SPCM0_1click_counts_iAll
    + SPCM1_1click_counts_iAll
    + SPCM0_OtherNode_1click_counts_iAll
    + SPCM1_OtherNode_1click_counts_iAll
)
AllSPCMs_2click_counts_iAll = (
    SPCM0_2click_counts_iAll
    + SPCM1_2click_counts_iAll
    + SPCM0_OtherNode_2click_counts_iAll
    + SPCM1_OtherNode_2click_counts_iAll
)


print(f'n_measurements = {n_measurements}')
print(f'excitation pulse width = {t_excitation_pulse/10**-9}','ns')
print(f'SPCM window = {t_photon_collection_time/10**-9}', 'ns')
print(f'excitation AOM power (dBm) = {p_excitation}')
print(f'n_excitation_attempts = {n_excitation_attempts}')
print(f'check atom avery n excitation cycles = {atom_check_every_n}')
if n_excitation_cycles_iAll[0] > 0:
    print(f'Overal photon detection efficiency (%) = {100*AllSPCMs_1click_counts_iAll[0]/n_excitation_cycles_iAll[0]}')
else:
    print('Overal photon detection efficiency (%) = nan')



#############  Average photon generation rate calculated from last timestamp minus first timestamp.
for i in range(iterations):
    if len(AllSPCMs_tStamps_filtered_iAll[i]) >= 2:
        t_duration_i = max(AllSPCMs_tStamps_filtered_iAll[i]) - min(AllSPCMs_tStamps_filtered_iAll[i])
        if t_duration_i > 0:
            Ave_photon_rate_iAll[i] = (AllSPCMs_1click_counts_iAll[i] + AllSPCMs_2click_counts_iAll[i]) / t_duration_i
        else:
            Ave_photon_rate_iAll[i] = np.nan
    else:
        Ave_photon_rate_iAll[i] = np.nan


###################################################### plot of 1-click and 2-click counts
fig, axs = plt.subplots(1, 5, figsize=(18, 3), sharey=True)

### SPCM0
axs[0].plot(scan_sequence1, SPCM0_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[0].plot(scan_sequence1, SPCM0_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[0].errorbar(scan_sequence1, SPCM0_1click_counts_iAll, yerr=np.sqrt(SPCM0_1click_counts_iAll), ls='none', color='tab:red')
axs[0].errorbar(scan_sequence1, SPCM0_2click_counts_iAll, yerr=np.sqrt(SPCM0_2click_counts_iAll), ls='none', color='tab:blue')
axs[0].set_title('SPCM0')
axs[0].set_ylabel('Occurrences')
axs[0].set_xlabel(scan_variable1_name);

### SPCM1
axs[1].plot(scan_sequence1, SPCM1_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[1].plot(scan_sequence1, SPCM1_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[1].errorbar(scan_sequence1, SPCM1_1click_counts_iAll, yerr=np.sqrt(SPCM1_1click_counts_iAll), ls='none', color='tab:red')
axs[1].errorbar(scan_sequence1, SPCM1_2click_counts_iAll, yerr=np.sqrt(SPCM1_2click_counts_iAll), ls='none', color='tab:blue')
axs[1].set_title('SPCM1')
axs[1].set_ylabel('Occurrences')
axs[1].set_xlabel(scan_variable1_name);

### SPCM0_OtherNode
axs[2].plot(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[2].plot(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[2].set_title('SPCM0_OtherNode')
axs[2].set_ylabel('Occurrences')
axs[2].set_xlabel(scan_variable1_name);

### SPCM1_OtherNode
axs[3].plot(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[3].plot(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[3].set_title('SPCM1_OtherNode')
axs[3].set_ylabel('Occurrences')
axs[3].set_xlabel(scan_variable1_name);

### all SPCMs
axs[4].plot(scan_sequence1, n_excitation_cycles_iAll * Col_Eff, label='Max expected', ls='solid', color='tab:gray', marker='o') ### I assumed only one excitation can happen with n_excitation_attempts
axs[4].plot(scan_sequence1, AllSPCMs_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[4].plot(scan_sequence1, AllSPCMs_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[4].errorbar(scan_sequence1, AllSPCMs_1click_counts_iAll, yerr=np.sqrt(AllSPCMs_1click_counts_iAll), ls='none', color='tab:red')
axs[4].errorbar(scan_sequence1, AllSPCMs_2click_counts_iAll, yerr=np.sqrt(AllSPCMs_2click_counts_iAll), ls='none', color='tab:blue')
axs[4].set_title('All SPCMs')
axs[4].set_ylabel('Occurrences')
axs[4].set_xlabel(scan_variable1_name);
axs[4].legend(loc=(1.05,0.2))
##########################################################################################




###################################################### plot of 2-click counts
fig, axs = plt.subplots(1, 5, figsize=(18, 3), sharey=True)

### SPCM0
# axs[0].plot(scan_sequence1, SPCM0_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[0].plot(scan_sequence1, SPCM0_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[0].errorbar(scan_sequence1, SPCM0_1click_counts_iAll, yerr=np.sqrt(SPCM0_1click_counts_iAll), ls='none', color='tab:red')
axs[0].errorbar(scan_sequence1, SPCM0_2click_counts_iAll, yerr=np.sqrt(SPCM0_2click_counts_iAll), ls='none', color='tab:blue')
axs[0].set_title('SPCM0')
axs[0].set_ylabel('Occurrences')
axs[0].set_xlabel(scan_variable1_name);

### SPCM1
# axs[1].plot(scan_sequence1, SPCM1_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[1].plot(scan_sequence1, SPCM1_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[1].errorbar(scan_sequence1, SPCM1_1click_counts_iAll, yerr=np.sqrt(SPCM1_1click_counts_iAll), ls='none', color='tab:red')
axs[1].errorbar(scan_sequence1, SPCM1_2click_counts_iAll, yerr=np.sqrt(SPCM1_2click_counts_iAll), ls='none', color='tab:blue')
axs[1].set_title('SPCM1')
axs[1].set_ylabel('Occurrences')
axs[1].set_xlabel(scan_variable1_name);

### SPCM0_OtherNode
# axs[2].plot(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[2].plot(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[2].set_title('SPCM0_OtherNode')
axs[2].set_ylabel('Occurrences')
axs[2].set_xlabel(scan_variable1_name);

### SPCM1_OtherNode
# axs[3].plot(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[3].plot(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[3].set_title('SPCM1_OtherNode')
axs[3].set_ylabel('Occurrences')
axs[3].set_xlabel(scan_variable1_name);

### all SPCMs
# axs[4].plot(scan_sequence1, n_excitation_cycles_iAll * Col_Eff, label='Max expected', ls='solid', color='tab:gray', marker='o') ### I assumed only one excitation can happen with n_excitation_attempts
# axs[4].plot(scan_sequence1, AllSPCMs_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[4].plot(scan_sequence1, AllSPCMs_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[4].errorbar(scan_sequence1, AllSPCMs_1click_counts_iAll, yerr=np.sqrt(AllSPCMs_1click_counts_iAll), ls='none', color='tab:red')
axs[4].errorbar(scan_sequence1, AllSPCMs_2click_counts_iAll, yerr=np.sqrt(AllSPCMs_2click_counts_iAll), ls='none', color='tab:blue')
axs[4].set_title('All SPCMs')
axs[4].set_ylabel('Occurrences')
axs[4].set_xlabel(scan_variable1_name);
axs[4].legend(loc=(1.05,0.2))
##########################################################################################


AllSPCMs_rel_tStamps_iAll = [
    np.concatenate((spcm0, spcm1, spcm0_other, spcm1_other))
    for spcm0, spcm1, spcm0_other, spcm1_other in zip(
        SPCM0_rel_tStamps_iAll,
        SPCM1_rel_tStamps_iAll,
        SPCM0_OtherNode_rel_tStamps_iAll,
        SPCM1_OtherNode_rel_tStamps_iAll,
    )
]

###################################################### plot of photon histograms
if show_photon_hist:
    for i in range(iterations):
        fig, axs = plt.subplots(1, 5, figsize=(18, 3), sharey=True, sharex=True)
    
        axs[0].hist(SPCM0_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[0].set_xlabel("detection time (ns)")
        axs[0].set_ylabel("Occurence")
        axs[0].set_title("SPCM0 photons")
    
        axs[1].hist(SPCM1_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[1].set_xlabel("detection time (ns)")
        axs[1].set_title("SPCM1 photons");
    
        axs[2].hist(SPCM0_OtherNode_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[2].set_xlabel("detection time (ns)")
        axs[2].set_title("SPCM0_OtherNode photons")
    
        axs[3].hist(SPCM1_OtherNode_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[3].set_xlabel("detection time (ns)")
        axs[3].set_title("SPCM1_OtherNode photons");
    
        axs[4].hist(AllSPCMs_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7, label=f'{scan_variable1_name}: {scan_sequence1[i]}')
        axs[4].set_xlabel("detection time (ns)")
        axs[4].set_title("All SPCMs photons")
        axs[4].legend(loc=(1.05, 0.2))
##########################################################################################


fig, axs = plt.subplots(1, 3, figsize=(12, 3))
########################################## Plot average number of excitation_cycles per measuremnt
axs[0].plot(scan_sequence1, n_excitation_cycles_iAll/n_measurements, label='ave excitation cycle', marker='o', color='tab:gray')
axs[0].set_xlabel(scan_variable1_name)
axs[0].set_ylabel('n_excitation_cycles')
axs[0].set_title('average n. of excitation cycles')

########################################## Plot g2
axs[1].plot(scan_sequence1, g2_0, marker='o', color='tab:red')
axs[1].errorbar(scan_sequence1, g2_0, yerr=g2_0_err, ls='none',color='tab:red')
axs[1].set_xlabel(scan_variable1_name)
# axs[1].set_ylabel('g2(0)')
axs[1].set_title('g2(0)');

########################################## Plot avergae rate
axs[2].plot(scan_sequence1, Ave_photon_rate_iAll, marker='o', color='tab:red')
# axs[2].legend(loc=(1.05,0.5))
axs[2].set_xlabel(scan_variable1_name)
axs[2].set_ylabel('rate (1/s)')
axs[2].set_title('average rate');

plt.subplots_adjust(wspace=0.4)  # Increase spacing between subplots
plt.show()
```

## combining files

### helper functions

```python
def get_full_path_from_index(file_index):
    f = fnames[file_index]
    return f if os.path.isabs(f) else os.path.join(results, f)


def read_h5_datasets(file_index):
    fname = get_full_path_from_index(file_index)

    data = {}
    with h5py.File(fname, "r") as h5:
        ds = h5["datasets"]
        for key in ds.keys():
            x = ds[key][()]
            if isinstance(x, bytes):
                x = x.decode()
            data[key] = x

    return data


def scalar_if_possible(x):
    x = np.asarray(x)
    if x.shape == ():
        return x.item()
    if x.size == 1:
        return x.ravel()[0].item()
    return x


def get_n_excitation_cycles_iAll(n_excitation_cycles, iterations, n_measurements):
    """
    Same logic as the working individual-file analysis:
        n_excitation_cycles_iAll[i] = sum(n_excitation_cycles[i*n_measurements+1:(i+1)*n_measurements+1] + 1)
    """
    n_excitation_cycles = np.asarray(n_excitation_cycles).ravel()
    n_excitation_cycles_iAll = np.zeros(iterations, dtype=int)

    for i in range(iterations):
        n_excitation_cycles_iAll[i] = np.sum(
            np.asarray(
                n_excitation_cycles[i*n_measurements+1:(i+1)*n_measurements+1],
                dtype=int,
            ) + 1
        )

    return n_excitation_cycles_iAll


def get_n_measurements_for_file(data, iterations):
    if "measurements" in data:
        return int(scalar_if_possible(data["measurements"]))

    if "n_measurements" in data:
        return int(scalar_if_possible(data["n_measurements"]))

    if "n_excitation_cycles" in data:
        return len(np.asarray(data["n_excitation_cycles"]).ravel()) // iterations

    raise KeyError("No measurements / n_measurements dataset found in this file.")


def get_n_excitation_cycles_iAll_from_data(data, iterations=None):
    """
    Return excitation cycles per scan point for THIS file.
    This avoids accidentally using a global n_excitation_cycles_iAll from another file.
    """
    if iterations is None:
        scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
        iterations = len(scan_sequence1)

    if "n_excitation_cycles_iAll" in data:
        return np.asarray(data["n_excitation_cycles_iAll"], dtype=int).ravel()

    if "n_excitation_cycles" in data:
        n_excitation_cycles = np.asarray(data["n_excitation_cycles"]).ravel()
        n_measurements_file = get_n_measurements_for_file(data, iterations)

        if n_excitation_cycles.size == 1:
            # Fallback for files that store only one constant value.
            return np.full(
                iterations,
                int(scalar_if_possible(n_excitation_cycles)) * n_measurements_file,
                dtype=int,
            )

        return get_n_excitation_cycles_iAll(
            n_excitation_cycles,
            iterations,
            n_measurements_file,
        )

    raise KeyError("No excitation-cycle dataset found in this file.")


def get_start_end_indices_for_file(data, n_excitation_cycles_iAll_file=None):
    """
    IMPORTANT FIX.

    Do NOT split the timestamp arrays into equal-sized chunks.
    Individual-file analysis uses:
        n_tStamps_iAll = n_excitation_cycles_iAll * n_excitation_attempts
        end_indices = np.cumsum(n_tStamps_iAll)
        start_indices = np.insert(end_indices[:-1], 0, 0)

    The accumulated-file analysis has to use the same boundaries.
    Otherwise photon timestamps from one scan point can be subtracted from
    reference timestamps belonging to a neighboring scan point, which creates
    fake out-of-window arrival times.
    """
    scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
    iterations = len(scan_sequence1)

    if n_excitation_cycles_iAll_file is None:
        n_excitation_cycles_iAll_file = get_n_excitation_cycles_iAll_from_data(
            data,
            iterations,
        )

    if "n_excitation_attempts" in data:
        n_excitation_attempts_file = int(scalar_if_possible(data["n_excitation_attempts"]))
    else:
        n_excitation_attempts_file = 1

    n_tStamps_iAll_file = np.asarray(n_excitation_cycles_iAll_file, dtype=int) * n_excitation_attempts_file

    end_indices = np.cumsum(n_tStamps_iAll_file)
    start_indices = np.insert(end_indices[:-1], 0, 0)

    n_total_reference = len(data["reference_tStamps_t1"]) - 1
    if int(end_indices[-1]) != int(n_total_reference):
        print(
            "WARNING: timestamp-boundary mismatch for this file. "
            f"sum(n_tStamps_iAll) = {int(end_indices[-1])}, "
            f"len(reference_tStamps_t1)-1 = {int(n_total_reference)}. "
            "Check n_excitation_cycles_iAll / n_excitation_cycles / n_excitation_attempts."
        )

    return start_indices, end_indices


def rel_timestamps_ns(tStamps_i, t1_reference_i):
    """
    Convert timestamp array for one scan point into relative detection times in ns.

    IMPORTANT:
    This now matches the individual-file analysis logic in get_rel_tStamps_i():
      - remove -1 no-click entries
      - subtract the matching t1 reference for that same shot
      - flatten 1-click and 2-click timestamps
      - remove negative relative timestamps

    The previous accumulated-file helper kept negative relative timestamps.
    That made the accumulated histogram look different from the individual-file histogram.
    """
    tStamps_i = np.asarray(tStamps_i)
    t1_reference_i = np.asarray(t1_reference_i)

    if tStamps_i.size == 0 or t1_reference_i.size == 0:
        return np.array([], dtype=float)

    # Make the shape shot x click_index.
    # For the normal max_clicks=2 case this is already 2D.
    if tStamps_i.ndim == 1:
        tStamps_i = tStamps_i.reshape(-1, 1)

    # Match the original individual-file logic, which uses zip(...)
    # and therefore only compares rows with existing references.
    n = min(len(tStamps_i), len(t1_reference_i))
    tStamps_i = tStamps_i[:n]
    t1_reference_i = t1_reference_i[:n]

    rel_chunks = [
        row[row != -1] - ref
        for row, ref in zip(tStamps_i, t1_reference_i)
    ]

    if len(rel_chunks) == 0:
        return np.array([], dtype=float)

    rel_tStamps_i = np.concatenate(rel_chunks) * 1e9
    rel_tStamps_i = rel_tStamps_i[np.isfinite(rel_tStamps_i)]

    # Same filter as the individual-file analysis:
    # in rare cases a photon timestamp is earlier than its reference.
    # Those were removed in the working individual-file path.
    rel_tStamps_i = rel_tStamps_i[rel_tStamps_i >= 0]

    return rel_tStamps_i


def first_click_exists(tStamps_i):
    tStamps_i = np.asarray(tStamps_i)

    if tStamps_i.ndim == 1:
        return tStamps_i != -1

    return tStamps_i[:, 0] != -1


def analyze_one_file(file_index):
    data = read_h5_datasets(file_index)

    scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
    iterations = len(scan_sequence1)

    # Get excitation cycles first, then use them to compute the SAME
    # start/end indices as the working individual-file analysis.
    n_excitation_cycles_iAll_file = get_n_excitation_cycles_iAll_from_data(data, iterations)
    start_indices, end_indices = get_start_end_indices_for_file(data, n_excitation_cycles_iAll_file)

    if "Atom_loading_time" in data:
        all_atom_loading_times = [np.asarray(data["Atom_loading_time"]).ravel()[1:]]
    else:
        all_atom_loading_times = []

    SPCM0_rel_tStamps_iAll = []
    SPCM1_rel_tStamps_iAll = []
    SPCM0_OtherNode_rel_tStamps_iAll = []
    SPCM1_OtherNode_rel_tStamps_iAll = []
    AllSPCMs_rel_tStamps_iAll = []

    H1H2_clicks_iAll = np.array([])
    V1V2_clicks_iAll = np.array([])
    H1V2_clicks_iAll = np.array([])
    V1H2_clicks_iAll = np.array([])

    H1V1_clicks_iAll = np.array([])
    H2V2_clicks_iAll = np.array([])

    H1orV1_and_H2orV2_clicks_iAll = np.array([])

    for i in range(iterations):
        start = start_indices[i]
        end = end_indices[i]

        SPCM0_tStamps_i = get_tStamps_i(
            data["SPCM0_Photon_tStamps"],
            start,
            end,
        )  # H1

        SPCM1_tStamps_i = get_tStamps_i(
            data["SPCM1_Photon_tStamps"],
            start,
            end,
        )  # V1

        SPCM0_OtherNode_tStamps_i = get_tStamps_i(
            data["SPCM0_OtherNode_Photon_tStamps"],
            start,
            end,
        )  # H2

        SPCM1_OtherNode_tStamps_i = get_tStamps_i(
            data["SPCM1_OtherNode_Photon_tStamps"],
            start,
            end,
        )  # V2

        t1_reference_i = data["reference_tStamps_t1"][1:][start:end]

        # Compute the relative timestamp arrays detector-by-detector.
        # Then build AllSPCMs_rel_tStamps_iAll from these four arrays.
        SPCM0_rel_i = rel_timestamps_ns(SPCM0_tStamps_i, t1_reference_i)
        SPCM1_rel_i = rel_timestamps_ns(SPCM1_tStamps_i, t1_reference_i)
        SPCM0_OtherNode_rel_i = rel_timestamps_ns(SPCM0_OtherNode_tStamps_i, t1_reference_i)
        SPCM1_OtherNode_rel_i = rel_timestamps_ns(SPCM1_OtherNode_tStamps_i, t1_reference_i)

        SPCM0_rel_tStamps_iAll.append(SPCM0_rel_i)
        SPCM1_rel_tStamps_iAll.append(SPCM1_rel_i)
        SPCM0_OtherNode_rel_tStamps_iAll.append(SPCM0_OtherNode_rel_i)
        SPCM1_OtherNode_rel_tStamps_iAll.append(SPCM1_OtherNode_rel_i)

        AllSPCMs_rel_tStamps_iAll.append(
            np.concatenate((
                SPCM0_rel_i,
                SPCM1_rel_i,
                SPCM0_OtherNode_rel_i,
                SPCM1_OtherNode_rel_i,
            ))
        )

        H1 = first_click_exists(SPCM0_tStamps_i)
        V1 = first_click_exists(SPCM1_tStamps_i)
        H2 = first_click_exists(SPCM0_OtherNode_tStamps_i)
        V2 = first_click_exists(SPCM1_OtherNode_tStamps_i)

        n = min(len(H1), len(V1), len(H2), len(V2))
        H1, V1, H2, V2 = H1[:n], V1[:n], H2[:n], V2[:n]

        H1H2_clicks_iAll = np.append(H1H2_clicks_iAll, np.sum(H1 & H2))
        V1V2_clicks_iAll = np.append(V1V2_clicks_iAll, np.sum(V1 & V2))
        H1V2_clicks_iAll = np.append(H1V2_clicks_iAll, np.sum(H1 & V2))
        V1H2_clicks_iAll = np.append(V1H2_clicks_iAll, np.sum(V1 & H2))

        H1V1_clicks_iAll = np.append(H1V1_clicks_iAll, np.sum(H1 & V1))
        H2V2_clicks_iAll = np.append(H2V2_clicks_iAll, np.sum(H2 & V2))

        H1orV1_and_H2orV2_clicks_iAll = np.append(
            H1orV1_and_H2orV2_clicks_iAll,
            np.sum((H1 | V1) & (H2 | V2)),
        )

    return {
        "file_index": file_index,
        "scan_sequence1": scan_sequence1,
        "scan_variable1_name": data["scan_variable1_name"],

        "n_excitation_cycles_iAll": n_excitation_cycles_iAll_file,

        "SPCM0_rel_tStamps_iAll": SPCM0_rel_tStamps_iAll,
        "SPCM1_rel_tStamps_iAll": SPCM1_rel_tStamps_iAll,
        "SPCM0_OtherNode_rel_tStamps_iAll": SPCM0_OtherNode_rel_tStamps_iAll,
        "SPCM1_OtherNode_rel_tStamps_iAll": SPCM1_OtherNode_rel_tStamps_iAll,
        "AllSPCMs_rel_tStamps_iAll": AllSPCMs_rel_tStamps_iAll,

        "H1H2": H1H2_clicks_iAll,
        "V1V2": V1V2_clicks_iAll,
        "H1V2": H1V2_clicks_iAll,
        "V1H2": V1H2_clicks_iAll,

        "H1V1": H1V1_clicks_iAll,
        "H2V2": H2V2_clicks_iAll,

        "H1orV1_and_H2orV2": H1orV1_and_H2orV2_clicks_iAll,

        "all_atom_loading_times": all_atom_loading_times,
    }


import re
from IPython.display import display, HTML


def get_rid_from_file(file_index):
    fname = get_full_path_from_index(file_index)

    rid = None

    with h5py.File(fname, "r") as h5:
        # Try h5 attributes first
        for key in ["rid", "RID", "run_id"]:
            if key in h5.attrs:
                rid = h5.attrs[key]
                break

        # Try datasets group
        if rid is None and "datasets" in h5:
            ds = h5["datasets"]
            for key in ["rid", "RID", "run_id"]:
                if key in ds:
                    rid = ds[key][()]
                    break

    if isinstance(rid, bytes):
        rid = rid.decode()

    if rid is not None:
        return scalar_if_possible(rid)

    # Fallback: try to extract RID from filename
    base = os.path.basename(fname)
    match = re.search(r"(?:RID|rid)[_\-\s]*(\d+)", base)

    if match is None:
        match = re.search(r"(\d{4,})", base)

    return match.group(1) if match else "unknown"


def get_file_summary_row(file_index):
    data = read_h5_datasets(file_index)

    scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
    iterations = len(scan_sequence1)

    if "measurements" in data:
        measurements = scalar_if_possible(data["measurements"])
    elif "n_measurements" in data:
        measurements = scalar_if_possible(data["n_measurements"])
    else:
        measurements = "unknown"

    try:
        n_excitation_cycles_iAll_file = get_n_excitation_cycles_iAll_from_data(data, iterations)
        total_excitation_cycles = int(np.sum(n_excitation_cycles_iAll_file))
    except Exception:
        total_excitation_cycles = len(data["reference_tStamps_t1"]) - 1

    return {
        "file_index": file_index,
        "RID": get_rid_from_file(file_index),
        "# measurements": measurements,
        "# total excitation cycles": total_excitation_cycles,
    }


def display_selected_file_summary(selected_file_indices):
    rows = [get_file_summary_row(file_index) for file_index in selected_file_indices]

    numeric_measurements = [row["# measurements"] for row in rows if isinstance(row["# measurements"], (int, float, np.integer, np.floating))]
    total_measurements = np.sum(numeric_measurements) if len(numeric_measurements) else "unknown"
    total_excitation_cycles = np.sum([row["# total excitation cycles"] for row in rows])

    html = """
    <table>
        <tr>
            <th>file_index</th>
            <th>RID</th>
            <th># measurements</th>
            <th># total excitation cycles</th>
        </tr>
    """

    for row in rows:
        html += f"""
        <tr>
            <td>{row["file_index"]}</td>
            <td>{row["RID"]}</td>
            <td>{row["# measurements"]}</td>
            <td>{row["# total excitation cycles"]}</td>
        </tr>
        """

    html += f"""
        <tr>
            <td><b>TOTAL</b></td>
            <td></td>
            <td><b>{total_measurements}</b></td>
            <td><b>{total_excitation_cycles}</b></td>
        </tr>
    """

    html += "</table>"

    display(HTML(html))

    return rows
```

### select files and combine

```python
# ============================================================
# Select files to accumulate
# ============================================================

    #### ZOOMED IN
    # 222:''
    # 230:''
    # 231:''
    # 236:''
    # 239:'' #### terminated almost in the end;  ##underflow
    # 241:'' ####underflow
    
    # 243:''
    # 244:''
    # 245:''  ## possible sync problem
    # 246:'', ## possible sync problem
    # 247:'', ## possible sync problem


    # 248: ''  ### good from here

    ### after improvements in node1's atom temperature
    # 258:''
    # 260:''
    # 261:''



selected_file_indices = [
    # ##### 189 + np.array([-50,-10,-6, -4, -2, 0, 2, 4, 6, 10, 50])
    # 33, 36,
    # 60, 62, 63, 64, 
    # # 65, 67, 68  ## no photons
    # 79, 82, 83, 84, 85, 86,
    # 87, 88, 89,
    # 97, 98, 99,
    
    # 138,
    # 144, 145, 146, 147,    ### synchronization before loading;
    # 151, 152, 153, 154, 
    # 161, 162,   ## exc relock???
    
    167, 169   ## node2 reboot
    
    # ##### np.arange(-25,24,5)
    # # 222, 230, 231, 236, 243, 244,
    # 258, 260, 261,  ### improved node1's atom temperature
    # 262, 263, 268  ### modified sync function


    ####### np.arange(-90,91,15)
    # # 123, 136, 137, 148, 161, 162, 163,  #### ~2026-06-27 - node1 rate not good; some synchronization issue;
    # 201, 202, 204, 205, #### 2026-06-29 - node1 rate fixed; synchronization fixed;
    # 206, 209, 210, 211,
    # 212, 213, 214, 215, 
    # 216, 217, 220, 221,

]

selected_file_summary = display_selected_file_summary(selected_file_indices)

# ============================================================
# Analyze selected files
# ============================================================

file_results = [analyze_one_file(file_index) for file_index in selected_file_indices]

scan_sequence1 = file_results[0]["scan_sequence1"]
scan_variable1_name = file_results[0]["scan_variable1_name"]
iterations = len(scan_sequence1)

# Check that all selected files have the same scan sequence
for r in file_results[1:]:
    if len(r["scan_sequence1"]) != len(scan_sequence1) or not np.allclose(r["scan_sequence1"], scan_sequence1):
        raise ValueError(
            f"File index {r['file_index']} does not have the same scan_sequence1."
        )

# ============================================================
# Accumulate detector-specific relative timestamps first
# ============================================================

SPCM0_rel_tStamps_iAll = [
    np.concatenate([r["SPCM0_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

SPCM1_rel_tStamps_iAll = [
    np.concatenate([r["SPCM1_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

SPCM0_OtherNode_rel_tStamps_iAll = [
    np.concatenate([r["SPCM0_OtherNode_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

SPCM1_OtherNode_rel_tStamps_iAll = [
    np.concatenate([r["SPCM1_OtherNode_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

# Now build AllSPCMs_rel_tStamps_iAll in exactly the same way as the
# individual-file analysis.
AllSPCMs_rel_tStamps_iAll = [
    np.concatenate((spcm0, spcm1, spcm0_other, spcm1_other))
    for spcm0, spcm1, spcm0_other, spcm1_other in zip(
        SPCM0_rel_tStamps_iAll,
        SPCM1_rel_tStamps_iAll,
        SPCM0_OtherNode_rel_tStamps_iAll,
        SPCM1_OtherNode_rel_tStamps_iAll,
    )
]

# Diagnostic: this should now reflect only true relative timestamps.
for i, arr in enumerate(AllSPCMs_rel_tStamps_iAll):
    if len(arr) > 0:
        print(
            f"scan i={i}, scan value={scan_sequence1[i]}: "
            f"N={len(arr)}, min={np.min(arr):.3f} ns, max={np.max(arr):.3f} ns"
        )
    else:
        print(f"scan i={i}, scan value={scan_sequence1[i]}: N=0")

# ============================================================
# Accumulate coincidence counts
# ============================================================

H1H2_clicks_iAll = np.sum([r["H1H2"] for r in file_results], axis=0)
V1V2_clicks_iAll = np.sum([r["V1V2"] for r in file_results], axis=0)
H1V2_clicks_iAll = np.sum([r["H1V2"] for r in file_results], axis=0)
V1H2_clicks_iAll = np.sum([r["V1H2"] for r in file_results], axis=0)

H1V1_clicks_iAll = np.sum([r["H1V1"] for r in file_results], axis=0)
H2V2_clicks_iAll = np.sum([r["H2V2"] for r in file_results], axis=0)

H1orV1_and_H2orV2_clicks_iAll = np.sum(
    [r["H1orV1_and_H2orV2"] for r in file_results],
    axis=0,
)

H1H2_or_V1V2_clicks_iAll = H1H2_clicks_iAll + V1V2_clicks_iAll
H1V2_or_V1H2_clicks_iAll = H1V2_clicks_iAll + V1H2_clicks_iAll
H1V1_or_H2V2_clicks_iAll = H1V1_clicks_iAll + H2V2_clicks_iAll

# ============================================================
# Accumulate excitation cycles
# ============================================================

n_excitation_cycles_iAll_combined = np.sum(
    [r["n_excitation_cycles_iAll"] for r in file_results],
    axis=0,
)

# Keep this name also, because later plotting cells may expect it.
n_excitation_cycles_iAll = n_excitation_cycles_iAll_combined

# ============================================================
# Accumulate atom loading times, if present
# ============================================================

all_atom_loading_times = []
for r in file_results:
    all_atom_loading_times.extend(r.get("all_atom_loading_times", []))

print("Accumulated file indices:", selected_file_indices)
print("scan_sequence1:", scan_sequence1)
print("n_excitation_cycles_iAll_combined:", n_excitation_cycles_iAll_combined)
```

### Atom loading time analysis

```python
import math
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Plot Atom_loading_time values for selected files
# Max 7 plots per row
# ============================================================

plots_per_row = 5
n_files = len(selected_file_indices)

if n_files == 0:
    raise ValueError("selected_file_indices is empty.")

ncols = min(plots_per_row, n_files)
nrows = math.ceil(n_files / plots_per_row)

fig, axs = plt.subplots(
    nrows=nrows,
    ncols=ncols,
    figsize=(4 * ncols, 3.2 * nrows),
    squeeze=False
)

axs = axs.ravel()

for plot_i, file_index in enumerate(selected_file_indices):
    ax = axs[plot_i]

    # Analyze this file only
    result = analyze_one_file(file_index)

    all_atom_loading_times = result["all_atom_loading_times"]

    # ============================================================
    # Flatten atom loading times safely
    # This matches your working logic:
    # combined_atom_loading_times = [t for loading_times in all_atom_loading_times for t in loading_times]
    # ============================================================

    if len(all_atom_loading_times) == 0:
        combined_atom_loading_times = np.array([])
    else:
        first_item = all_atom_loading_times[0]

        # Case 1: all_atom_loading_times is a list of arrays
        if np.ndim(first_item) > 0:
            combined_atom_loading_times = np.concatenate([
                np.asarray(loading_times).ravel()
                for loading_times in all_atom_loading_times
            ])

        # Case 2: all_atom_loading_times is already one flat array/list
        else:
            combined_atom_loading_times = np.asarray(all_atom_loading_times).ravel()

    if len(combined_atom_loading_times) == 0:
        ax.set_title(f"File {file_index}: no loading times")
        ax.axis("off")
        continue

    mean_loading_time = np.mean(combined_atom_loading_times)

    ax.plot(combined_atom_loading_times, 'r-', label='Atom loading time')

    ax.set_xlabel("Atom number")
    ax.set_ylabel("Loading time (s)")
    ax.set_title(
        f"File {file_index}\nMean: {mean_loading_time:.3f} s"
    )
    ax.legend(loc='best')

    print(
        f"file {file_index}:",
        "N =", len(combined_atom_loading_times),
        "first 5 =", combined_atom_loading_times[:5],
        "mean =", mean_loading_time
    )

# Hide unused empty subplots
for j in range(n_files, len(axs)):
    axs[j].axis("off")

plt.tight_layout()
plt.show()
```

### HOM fitting

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit


# ============================================================
# Scaling counts to common excitation cycles
# ============================================================

def scale_to_common_cycles(counts, n_cycles, N_common=None):
    counts = np.asarray(counts, dtype=float)
    n_cycles = np.asarray(n_cycles, dtype=float)

    if N_common is None:
        N_common = np.mean(n_cycles)

    y_scaled = counts / n_cycles * N_common

    # Poisson error on raw counts, then scaled the same way.
    # Use max(counts, 1) so zero-count points do not get zero error.
    yerr_scaled = np.sqrt(np.maximum(counts, 1.0)) / n_cycles * N_common

    return y_scaled, yerr_scaled, N_common


# ============================================================
# Gaussian HOM dip model
# ============================================================

# def hom_gaussian_dip(x, y_far, depth, x0, sigma):
#     """
#     Gaussian HOM dip:

#         y(x) = y_far - depth * exp[-(x - x0)^2 / (2 sigma^2)]

#     y_far : coincidence level far away from temporal overlap
#     depth : dip depth
#     x0    : dip center
#     sigma : dip width
#     """
#     return y_far - depth * np.exp(-0.5 * ((x - x0) / sigma) ** 2)

def hom_gaussian_dip(x, y_far, visibility, x0, sigma):
    """
    Gaussian HOM dip constrained to stay non-negative:

        y(x) = y_far * [1 - visibility * exp(-(x - x0)^2 / (2 sigma^2))]

    visibility is bounded between 0 and 1.
    """
    return y_far * (1.0 - visibility * np.exp(-0.5 * ((x - x0) / sigma) ** 2))
    

def make_hom_initial_guess(x, y):
    """
    Automatically choose reasonable initial guesses from the data.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    good = np.isfinite(x) & np.isfinite(y)
    x = x[good]
    y = y[good]

    y_min = np.min(y)
    y_max = np.max(y)

    # Estimate far-away level from the high-count points.
    # This is more stable than using only max(y).
    n_top = max(2, int(np.ceil(0.3 * len(y))))
    y_sorted = np.sort(y)
    y_far_guess = np.mean(y_sorted[-n_top:])

    # Dip center guess: location of minimum count.
    x0_guess = x[np.argmin(y)]

    # Dip depth guess.
    depth_guess = max(y_far_guess - y_min, 0.1 * max(y_far_guess, 1.0))

    # Width guess: use scan range / 5.
    # x_range = np.max(x) - np.min(x)
    # sigma_guess = max(x_range / 5.0, 1.0)
    
    # Width guess: choose FWHM guess directly.
    fwhm_guess = 26.0  # ns
    sigma_guess = fwhm_guess / (2.0 * np.sqrt(2.0 * np.log(2.0)))

    return y_far_guess, depth_guess, x0_guess, sigma_guess


def fit_single_hom_dip(
    x,
    counts,
    n_cycles,
    label,
    N_common=None,
    x_label="Timing difference in ns",
    plot_residuals=True,
):
    """
    Scale one coincidence-count dataset, fit a Gaussian HOM dip,
    and plot the data separately with the fit.
    """

    x = np.asarray(x, dtype=float)
    counts = np.asarray(counts, dtype=float)
    n_cycles = np.asarray(n_cycles, dtype=float)

    y_scaled, yerr_scaled, N_common_used = scale_to_common_cycles(
        counts,
        n_cycles,
        N_common=N_common,
    )

    # Remove bad points
    good = (
        np.isfinite(x)
        & np.isfinite(y_scaled)
        & np.isfinite(yerr_scaled)
        & (n_cycles > 0)
    )

    x_fitdata = x[good]
    y_fitdata = y_scaled[good]
    yerr_fitdata = yerr_scaled[good]

    # Sort by x
    order = np.argsort(x_fitdata)
    x_fitdata = x_fitdata[order]
    y_fitdata = y_fitdata[order]
    yerr_fitdata = yerr_fitdata[order]

    if len(x_fitdata) < 4:
        raise ValueError(f"{label}: Need at least 4 valid data points for this fit.")

    # ============================================================
    # Adaptive initial guesses
    # ============================================================
    # p0 = make_hom_initial_guess(x_fitdata, y_fitdata)

    # y_far_guess, depth_guess, x0_guess, sigma_guess = p0

    y_far_guess, depth_guess, x0_guess, sigma_guess = make_hom_initial_guess(
        x_fitdata,
        y_fitdata,
    )
    
    visibility_guess = depth_guess / y_far_guess if y_far_guess > 0 else 0.5
    visibility_guess = np.clip(visibility_guess, 0.01, 0.99)
    
    p0 = [
        y_far_guess,
        visibility_guess,
        x0_guess,
        sigma_guess,
    ]
    
    # print("\nInitial guesses for", label)
    # print("----------------------------------------")
    # print(f"y_far guess = {y_far_guess:.4g}")
    # print(f"depth guess = {depth_guess:.4g}")
    # print(f"x0 guess    = {x0_guess:.4g} ns")
    # print(f"sigma guess = {sigma_guess:.4g} ns")

    print("\nInitial guesses for", label)
    print("----------------------------------------")
    print(f"y_far guess      = {y_far_guess:.4g}")
    print(f"visibility guess = {visibility_guess:.4g}")
    print(f"x0 guess         = {x0_guess:.4g} ns")
    print(f"sigma guess      = {sigma_guess:.4g} ns")

    # ============================================================
    # Bounds
    # ============================================================
    y_max = np.max(y_fitdata)
    x_min = np.min(x_fitdata)
    x_max = np.max(x_fitdata)
    x_range = x_max - x_min

    # lower_bounds = [
    #     0.0,          # y_far
    #     0.0,          # depth
    #     x_min,        # x0
    #     0.1,          # sigma
    # ]

    # upper_bounds = [
    #     5.0 * max(y_max, 1.0),       # y_far
    #     5.0 * max(y_max, 1.0),       # depth
    #     x_max,                       # x0
    #     5.0 * max(x_range, 1.0),     # sigma
    # ]

    lower_bounds = [
        0.0,          # y_far
        0.0,          # visibility
        x_min,        # x0
        0.1,          # sigma
    ]
    
    upper_bounds = [
        5.0 * max(y_max, 1.0),       # y_far
        1.0,                         # visibility cannot exceed 100%
        x_max,                       # x0
        5.0 * max(x_range, 1.0),     # sigma
    ]

    # ============================================================
    # Fit
    # ============================================================
    popt, pcov = curve_fit(
        hom_gaussian_dip,
        x_fitdata,
        y_fitdata,
        p0=p0,
        sigma=yerr_fitdata,
        absolute_sigma=True,
        bounds=(lower_bounds, upper_bounds),
        maxfev=50000,
    )

    perr = np.sqrt(np.diag(pcov))

    # y_far, depth, x0, sigma = popt
    # y_far_err, depth_err, x0_err, sigma_err = perr

    # # HOM visibility from Gaussian dip fit
    # visibility = depth / y_far if y_far != 0 else np.nan

    y_far, visibility, x0, sigma = popt
    y_far_err, visibility_err, x0_err, sigma_err = perr
    
    depth = y_far * visibility
    
    # Error propagation for depth = y_far * visibility
    depth_var = (
        visibility**2 * pcov[0, 0]
        + y_far**2 * pcov[1, 1]
        + 2 * y_far * visibility * pcov[0, 1]
    )
    depth_err = np.sqrt(max(depth_var, 0.0))

    # if depth > 0 and y_far > 0:
    #     visibility_err = visibility * np.sqrt(
    #         (depth_err / depth) ** 2 + (y_far_err / y_far) ** 2
    #     )
    # else:
    #     visibility_err = np.nan

    fwhm = 2.0 * np.sqrt(2.0 * np.log(2.0)) * sigma
    fwhm_err = 2.0 * np.sqrt(2.0 * np.log(2.0)) * sigma_err

    # Fit quality
    y_model_data = hom_gaussian_dip(x_fitdata, *popt)
    residuals = y_fitdata - y_model_data

    chi2 = np.sum((residuals / yerr_fitdata) ** 2)
    dof = len(x_fitdata) - len(popt)
    reduced_chi2 = chi2 / dof if dof > 0 else np.nan

    # Smooth fit curve
    x_smooth = np.linspace(x_min, x_max, 1000)
    y_smooth = hom_gaussian_dip(x_smooth, *popt)

    # ============================================================
    # Print fit result
    # ============================================================
    # print("\nFit result for", label)
    # print("============================================================")
    # print(f"y_far      = {y_far:.4g} ± {y_far_err:.4g}")
    # print(f"depth      = {depth:.4g} ± {depth_err:.4g}")
    # print(f"x0         = {x0:.4g} ± {x0_err:.4g} ns")
    # print(f"sigma      = {sigma:.4g} ± {sigma_err:.4g} ns")
    # print(f"FWHM       = {fwhm:.4g} ± {fwhm_err:.4g} ns")
    # print(f"visibility = {visibility:.4f} ± {visibility_err:.4f}")
    # print(f"reduced chi^2 = {reduced_chi2:.3g}")
    # print("============================================================")

    print("\nFit result for", label)
    print("============================================================")
    print(f"y_far      = {y_far:.4g} ± {y_far_err:.4g}")
    print(f"depth      = {depth:.4g} ± {depth_err:.4g}")
    print(f"x0         = {x0:.4g} ± {x0_err:.4g} ns")
    print(f"sigma      = {sigma:.4g} ± {sigma_err:.4g} ns")
    print(f"FWHM       = {fwhm:.4g} ± {fwhm_err:.4g} ns")
    print(f"visibility = {visibility:.4f} ± {visibility_err:.4f}")
    print(f"reduced chi^2 = {reduced_chi2:.3g}")
    print("============================================================")

    # ============================================================
    # Plot data and fit
    # ============================================================
    fig, ax = plt.subplots(1, 1, figsize=(8, 5))

    ax.errorbar(
        x_fitdata,
        y_fitdata,
        yerr=yerr_fitdata,
        fmt="o",
        capsize=3,
        label=label,
    )

    ax.plot(
        x_smooth,
        y_smooth,
        linewidth=2,
        label=(
            f"Gaussian HOM fit\n"
            f"$x_0$ = {x0:.2f} ns, "
            f"$V$ = {100 * visibility:.1f}%, "
            f"FWHM = {fwhm:.2f} ns"
        ),
    )

    ax.axvline(x0, linestyle="--", linewidth=1, label="fit center")

    ax.set_xlabel(x_label)
    ax.set_ylabel(f"equivalent counts for {N_common_used:.0f} excitation cycles")
    ax.set_title(f"HOM dip fit: {label}")
    ax.legend()
    ax.grid(alpha=0.3)

    plt.tight_layout()
    plt.show()

    # ============================================================
    # Residual plot
    # ============================================================
    if plot_residuals:
        fig, ax = plt.subplots(1, 1, figsize=(8, 3))

        ax.errorbar(
            x_fitdata,
            residuals,
            yerr=yerr_fitdata,
            fmt="o",
            capsize=3,
        )

        ax.axhline(0, linestyle="--", linewidth=1)

        ax.set_xlabel(x_label)
        ax.set_ylabel("residual")
        ax.set_title(f"Fit residuals: {label}")
        ax.grid(alpha=0.3)

        plt.tight_layout()
        plt.show()

    return {
        "label": label,
        "x": x_fitdata,
        "y_scaled": y_fitdata,
        "yerr_scaled": yerr_fitdata,
        "N_common": N_common_used,
        "popt": popt,
        "perr": perr,
        "y_far": y_far,
        "y_far_err": y_far_err,
        "depth": depth,
        "depth_err": depth_err,
        "x0": x0,
        "x0_err": x0_err,
        "sigma": sigma,
        "sigma_err": sigma_err,
        "fwhm": fwhm,
        "fwhm_err": fwhm_err,
        "visibility": visibility,
        "visibility_err": visibility_err,
        "chi2": chi2,
        "dof": dof,
        "reduced_chi2": reduced_chi2,
        "x_smooth": x_smooth,
        "y_smooth": y_smooth,
        "residuals": residuals,
    }


# ============================================================
# Run fits separately for your two datasets
# ============================================================

x_scan = np.asarray(scan_sequence1, dtype=float) - 189
N_common = np.mean(n_excitation_cycles_iAll_combined)

# hom_fit_all_HV = fit_single_hom_dip(
#     x=x_scan,
#     counts=H1orV1_and_H2orV2_clicks_iAll,
#     n_cycles=n_excitation_cycles_iAll_combined,
#     label="(H1 or V1) and (H2 or V2)",
#     N_common=N_common,
# )

hom_fit_HH_plus_VV = fit_single_hom_dip(
    x=x_scan,
    counts=H1H2_or_V1V2_clicks_iAll,
    n_cycles=n_excitation_cycles_iAll_combined,
    label="H1H2 + V1V2",
    N_common=N_common,
)
```

```python
"""
What each variable represents. Example values are mostly from RID 22496, 2025-02-26. Variables ending in _iAll contain data for all interations with length equal to "iterations".

n_measurements = 2000: directly from dataset
iterations = 5: the length of scan_sequence1, directly from dataset
n_excitation_cycles = [0,  2, 20,  8,  2,  8, ...], length = 1 + n_measurements * iterations: the number of excitation cycles in each measurement
n_excitation_cycles_iAll = [16830, 16665, 16488, 16194, 16221], length = iterations: total number of excitation cycles in each iteration as an array.
n_excitation_attempts = 1: directly from dataset. Number of excitation pulses in each excitation cycle.
n_tStamps_iAll = [16830, 16665, 16488, 16194, 16221], length = iterations: total number of timestamps in each iteration as an array.
SPCM0_Photon_tStamps = [[0, 0], [-1, -1], [-1, -1], [1.32237714e+06, -1], [-1, -1], ...], length = sum(n_tStamps_iAll) = 82399, for example. Directly from dataset.
SPCM0_tStamps_i =  [[-1, -1], [-1, -1], [1.32237714e+06, -1], [-1, -1], ...], length = elements of n_tStamps_iAll, 16830 for the first iteration, for example. The timestamps in each iteration.
reference_tStamps_t1 = [0, 1322341.65973562, 1322341.65976682, 1322341.65979802, ...], length = sum(n_tStamps_iAll) = 82399, for example. Directly from dataset.
t1_reference_i = [1322341.65973562, 1322341.65976682, 1322341.65979802, ...], length = elements of n_tStamps_iAll, 16830 for the first iteration, for example. The reference times in each iteration.
SPCM0_1click_counts_iAll = [128., 122., 172., 186., 172.], length = iterations: number of 1-click events in each iteration as an array; events in the first column of SPCM0_tStamps_i for different iteration.
SPCM0_2click_counts_iAll = [0., 0., 0., 0., 0.], length = iterations: number of 2-click events in each iteration as an array; events in the second column of SPCM0_tStamps_i for different iteration.
start_indices = [0, 16830, 33495, 49983, 66177], length = iterations: to slice arrays like SPCM0_Photon_tStamps into iterations.
    Remember that SPCM0_Photon_tStamps is a single array with the timestamps for all iterations.
end_indices = [16830, 33495, 49983, 66177, 82398], length = iterations: same as start_indices
SPCM0_rel_tStamps_i = [1099.89196062, 1088.94892037, 1098.02931547, ...], length = SPCM0_1click_counts_iAll + SPCM0_2click_counts_iAll, for example length is 128 and 122 for first two iterations.
    This array contains the timestamps relative to the reference time t1 in ns unit. -1 event are excluded and 2-click events are included.
SPCM0_rel_tStamps_iAll = [[1099.89196062, 1088.94892037, 1098.02931547, ...], [..], ..], length = iterations. Arrays of arrays containing all the relative timestamps for all iteration.
AllSPCMs_tStamps_filtered_i = [[1757511.19383094, 1757533.11171639, ...], length = total number of photons in iteration i,
    i.e. length = all 1-click and 2-click counts from SPCM0, SPCM1, SPCM0_OtherNode, and SPCM1_OtherNode. Contains all the absolute time stamps from all 4 SPCMs with -1 removed and sorted from small to large.
AllSPCMs_tStamps_filtered_iAll = [[1757511.19383094, 1757533.11171639, ...], [1749316.29104084, 1749328.33163165, ..], ..], length = iterations. Array of arrays keeping all
    AllSPCMs_tStamps_filtered_i from all iterations.
Ave_photon_rate_iAll = [[0.11084697, 0.1147433 , 0.10952382, 0.1337287 , 0.12123424]], length = iterations. Contains the average photon generation rate for all iterations.
n_coincidence_iAll = [0,0,1,0,2], length = iterations. The number of coincidences in each iteration used to calculate g2.

Other notes:
1- Does this work if I change max_clicks from 2 to 1 or 3? The answer is no.

"""
show_photon_hist = False
### Print the title of the output
display(HTML(f"""<b style='font-size:18px;'> {date_filters[0]} - RID {rid} - {experiment_function}</b>"""))

Col_Eff = 0.05 # overal expected/theoretical collection and detection efficiency

### Instead of using a large n_measurements, which can sometimes lead to underflow error, we use dummy_variable to split the measurements.
### So, if scan_variable1_name == 'dummy_variable', we want to combine all iterations into one:
if scan_variable1_name == 'dummy_variable':
    n_measurements = n_measurements * iterations
    iterations = 1
    scan_sequence1 = [0]

## To look at specific iterations, you can select in scan_sequence_1, for example scan_sequence1 = [0,2]. For now, the first element has to be 0.
# scan_sequence1 = [0,4]
# iterations = len(scan_sequence1)
# n_measurements = n_measurements * iterations


n_excitation_cycles_iAll = np.zeros(iterations, dtype=int) ### total number of excitation cycle in each iteration.
n_tStamps_iAll = np.zeros(iterations, dtype=int)  ### number of timestamps in each iteration.
for i in range(iterations):
    n_excitation_cycles_iAll[i] = sum(n_excitation_cycles[i*n_measurements+1:(i+1)*n_measurements+1]+1)
    if scan_variable1_name == 'n_excitation_attempts':
        n_excitation_attempts = scan_sequence1[i]
    n_tStamps_iAll[i] = n_excitation_cycles_iAll[i] * n_excitation_attempts
    ### Note that n_excitation_cycles varies in each cycle, each contains n_excitation_attempts.
    ### the first +1 is because the first element in each dataset is 0.
    ### the last +1 is because there is one more cycle than n_excitation_cycles in each loop. See the oscciloscope. For example, if n_excitation_cycles[j] = 2,
    ### there are 2+1=3 sets of excitation cycles for that measurement.


SPCM0_1click_counts_iAll = np.array([])
SPCM0_2click_counts_iAll = np.array([])
SPCM1_1click_counts_iAll = np.array([])
SPCM1_2click_counts_iAll = np.array([])
SPCM0_OtherNode_1click_counts_iAll = np.array([])
SPCM0_OtherNode_2click_counts_iAll = np.array([])
SPCM1_OtherNode_1click_counts_iAll = np.array([])
SPCM1_OtherNode_2click_counts_iAll = np.array([])

AllSPCMs_tStamps_filtered_iAll = []

SPCM0_rel_tStamps_iAll = []
SPCM1_rel_tStamps_iAll = []
SPCM0_OtherNode_rel_tStamps_iAll = []
SPCM1_OtherNode_rel_tStamps_iAll = []

n_coincidence_iAll = np.zeros(iterations)
g2_0 = np.zeros(iterations)
g2_0_err = np.zeros(iterations)
Ave_photon_rate_iAll = np.zeros(iterations)


def get_rel_tStamps_i(tStamps_i, t1_reference_i, spcm_name):
    ### the timestamps relative to reference time. -1 events are excluded.
    rel_tStamps_i_temp = [row[row != -1] - ref for row, ref in zip(tStamps_i, t1_reference_i)]
    if not rel_tStamps_i_temp: ### np.concatenate crashes on empty lists; if no photon detected, I replace timetags by [0].
        rel_tStamps_i_temp = [[0]]
    rel_tStamps_i_temp2 = np.concatenate(rel_tStamps_i_temp) * 1e9 ## Flatten to combine 1-click and 2-click events.
    rel_tStamps_i = rel_tStamps_i_temp2[rel_tStamps_i_temp2 >= 0]
    ## in 1 out of 1million case, the reference time is later than the timetag and I get a negative
    ### element which messes up the histograms. I am simply removing these elements if any, and printing
    if len(rel_tStamps_i) != len(rel_tStamps_i_temp2):
        print(f'Early time stamp for {len(rel_tStamps_i_temp2) - len(rel_tStamps_i)} elements on {spcm_name} was detected.')
    return rel_tStamps_i


def get_tStamps_i(tStamps_all, start, end):
    tStamps_i = np.asarray(tStamps_all[1:][start:end])
    if tStamps_i.size == 0:
        return np.empty((0, 2))
    if tStamps_i.ndim == 1:
        return tStamps_i.reshape(-1, 2)
    return tStamps_i


### Calculate the ending and starting indices for the iterations
end_indices = np.cumsum(n_tStamps_iAll)
start_indices = np.insert(end_indices[:-1], 0, 0) ### prepend a 0 for the starting index

for i in range(iterations):
    start = start_indices[i]
    end = end_indices[i]
    SPCM0_tStamps_i = get_tStamps_i(SPCM0_Photon_tStamps, start, end)
    SPCM1_tStamps_i = get_tStamps_i(SPCM1_Photon_tStamps, start, end)
    SPCM0_OtherNode_tStamps_i = get_tStamps_i(SPCM0_OtherNode_Photon_tStamps, start, end)
    SPCM1_OtherNode_tStamps_i = get_tStamps_i(SPCM1_OtherNode_Photon_tStamps, start, end)
    t1_reference_i = reference_tStamps_t1[1:][start:end]

    ### number of 1-click events (the first column is none -1)
    SPCM0_1click_counts_iAll = np.append(SPCM0_1click_counts_iAll, np.count_nonzero(SPCM0_tStamps_i[:, 0] != -1))
    SPCM1_1click_counts_iAll = np.append(SPCM1_1click_counts_iAll, np.count_nonzero(SPCM1_tStamps_i[:, 0] != -1))
    SPCM0_OtherNode_1click_counts_iAll = np.append(SPCM0_OtherNode_1click_counts_iAll, np.count_nonzero(SPCM0_OtherNode_tStamps_i[:, 0] != -1))
    SPCM1_OtherNode_1click_counts_iAll = np.append(SPCM1_OtherNode_1click_counts_iAll, np.count_nonzero(SPCM1_OtherNode_tStamps_i[:, 0] != -1))

    ### number of 2-click events (the 2nd column is none -1)
    SPCM0_2click_counts_iAll = np.append(SPCM0_2click_counts_iAll, np.count_nonzero(SPCM0_tStamps_i[:, 1] != -1))
    SPCM1_2click_counts_iAll = np.append(SPCM1_2click_counts_iAll, np.count_nonzero(SPCM1_tStamps_i[:, 1] != -1))
    SPCM0_OtherNode_2click_counts_iAll = np.append(SPCM0_OtherNode_2click_counts_iAll, np.count_nonzero(SPCM0_OtherNode_tStamps_i[:, 1] != -1))
    SPCM1_OtherNode_2click_counts_iAll = np.append(SPCM1_OtherNode_2click_counts_iAll, np.count_nonzero(SPCM1_OtherNode_tStamps_i[:, 1] != -1))

    SPCM0_rel_tStamps_i = get_rel_tStamps_i(SPCM0_tStamps_i, t1_reference_i, 'SPCM0')
    SPCM1_rel_tStamps_i = get_rel_tStamps_i(SPCM1_tStamps_i, t1_reference_i, 'SPCM1')
    SPCM0_OtherNode_rel_tStamps_i = get_rel_tStamps_i(SPCM0_OtherNode_tStamps_i, t1_reference_i, 'SPCM0_OtherNode')
    SPCM1_OtherNode_rel_tStamps_i = get_rel_tStamps_i(SPCM1_OtherNode_tStamps_i, t1_reference_i, 'SPCM1_OtherNode')

    SPCM0_rel_tStamps_iAll.append(SPCM0_rel_tStamps_i)
    SPCM1_rel_tStamps_iAll.append(SPCM1_rel_tStamps_i)
    SPCM0_OtherNode_rel_tStamps_iAll.append(SPCM0_OtherNode_rel_tStamps_i)
    SPCM1_OtherNode_rel_tStamps_iAll.append(SPCM1_OtherNode_rel_tStamps_i)

    ### calculate g2 for each iteration. In future, we can also select a narrower coincidence window.
    ### For 4 SPCMs, count coincidences between all unordered detector pairs.
    first_click_arrays_i = [
        SPCM0_tStamps_i[:, 0],
        SPCM1_tStamps_i[:, 0],
        SPCM0_OtherNode_tStamps_i[:, 0],
        SPCM1_OtherNode_tStamps_i[:, 0],
    ]
    first_click_counts_i = [
        SPCM0_1click_counts_iAll[i],
        SPCM1_1click_counts_iAll[i],
        SPCM0_OtherNode_1click_counts_iAll[i],
        SPCM1_OtherNode_1click_counts_iAll[i],
    ]

    n_coincidence_i = 0
    g2_denominator_i = 0
    g2_denominator_var_i = 0
    for j in range(len(first_click_arrays_i)):
        for k in range(j + 1, len(first_click_arrays_i)):
            n_coincidence_i += np.sum((first_click_arrays_i[j] >= 1) & (first_click_arrays_i[k] >= 1))

            Nj = first_click_counts_i[j]
            Nk = first_click_counts_i[k]
            g2_denominator_i += Nj * Nk
            g2_denominator_var_i += (Nk**2) * Nj + (Nj**2) * Nk

    # n_coincidence_i = 1 ### if there is no coincidence, you can set n_coincidence_i = 1 manually to find the bound to g2 because the next count might be a coincidence.
    n_coincidence_iAll[i] = n_coincidence_i ### Having a list of coincidences for each iteration helps to learn more about the g2.

    if g2_denominator_i > 0:
        g2_0_temp = (n_coincidence_i / g2_denominator_i) * n_excitation_cycles_iAll[i]
        ### Errors for g2 results are from error propagation assuming Poissonian distribution for counts:
        if n_coincidence_i > 0:
            g2_0_err_temp = g2_0_temp * np.sqrt(1/n_coincidence_i + g2_denominator_var_i/(g2_denominator_i**2))
        else:
            g2_0_err_temp = np.nan
    else:
        g2_0_temp = np.nan
        g2_0_err_temp = np.nan
    g2_0[i] = g2_0_temp
    g2_0_err[i] = g2_0_err_temp

    ############## absolute timestamps of all the photons for rate analysis
    AllSPCMs_tStamps_flatten_i = np.concatenate((
        SPCM0_tStamps_i,
        SPCM1_tStamps_i,
        SPCM0_OtherNode_tStamps_i,
        SPCM1_OtherNode_tStamps_i,
    )).flatten() ### flatten to get the timestamps of all photons (1st and 2nd clicks).
    AllSPCMs_tStamps_filtered_i = AllSPCMs_tStamps_flatten_i[AllSPCMs_tStamps_flatten_i != -1]  ### -1 events are removed
    AllSPCMs_tStamps_filtered_i = np.sort(AllSPCMs_tStamps_filtered_i) ### sorted from small to large timestamps
    AllSPCMs_tStamps_filtered_iAll.append(AllSPCMs_tStamps_filtered_i)


    # ###########checking H1H2 ###############
    # count=0
    # for j in range(len(SPCM0_tStamps_i)):
    #     if SPCM0_tStamps_i[:, 0][j] != -1 and SPCM0_OtherNode_tStamps_i[:, 0][j] != -1:
    #         count+=1
    
    # print("H1H2 ", count)
    # ###########checking V1V2 ###############
    # count=0
    # for j in range(len(SPCM1_tStamps_i)):
    #     if SPCM1_tStamps_i[:, 0][j] != -1 and SPCM1_OtherNode_tStamps_i[:, 0][j] != -1:
    #         count+=1
    
    # print("V1V2 ",count)
    
    # ###########checking (H1orV1) (H2orV2) ###############
    # count=0
    # for j in range(len(SPCM1_tStamps_i)):
    #     if (SPCM0_tStamps_i[:, 0][j] != -1 or SPCM1_tStamps_i[:, 0][j] != -1) and (SPCM0_OtherNode_tStamps_i[:, 0][j] != -1 or SPCM1_OtherNode_tStamps_i[:, 0][j] != -1):
    #         count+=1
    
    # print("(H1orV1) (H2orV2) ",count)
    


AllSPCMs_1click_counts_iAll = (
    SPCM0_1click_counts_iAll
    + SPCM1_1click_counts_iAll
    + SPCM0_OtherNode_1click_counts_iAll
    + SPCM1_OtherNode_1click_counts_iAll
)
AllSPCMs_2click_counts_iAll = (
    SPCM0_2click_counts_iAll
    + SPCM1_2click_counts_iAll
    + SPCM0_OtherNode_2click_counts_iAll
    + SPCM1_OtherNode_2click_counts_iAll
)


print(f'n_measurements = {n_measurements}')
print(f'excitation pulse width = {t_excitation_pulse/10**-9}','ns')
print(f'SPCM window = {t_photon_collection_time/10**-9}', 'ns')
print(f'excitation AOM power (dBm) = {p_excitation}')
print(f'n_excitation_attempts = {n_excitation_attempts}')
print(f'check atom avery n excitation cycles = {atom_check_every_n}')
if n_excitation_cycles_iAll[0] > 0:
    print(f'Overal photon detection efficiency (%) = {100*AllSPCMs_1click_counts_iAll[0]/n_excitation_cycles_iAll[0]}')
else:
    print('Overal photon detection efficiency (%) = nan')



#############  Average photon generation rate calculated from last timestamp minus first timestamp.
for i in range(iterations):
    if len(AllSPCMs_tStamps_filtered_iAll[i]) >= 2:
        t_duration_i = max(AllSPCMs_tStamps_filtered_iAll[i]) - min(AllSPCMs_tStamps_filtered_iAll[i])
        if t_duration_i > 0:
            Ave_photon_rate_iAll[i] = (AllSPCMs_1click_counts_iAll[i] + AllSPCMs_2click_counts_iAll[i]) / t_duration_i
        else:
            Ave_photon_rate_iAll[i] = np.nan
    else:
        Ave_photon_rate_iAll[i] = np.nan


###################################################### plot of 1-click and 2-click counts
fig, axs = plt.subplots(1, 5, figsize=(18, 3), sharey=True)

### SPCM0
axs[0].plot(scan_sequence1, SPCM0_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[0].plot(scan_sequence1, SPCM0_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[0].errorbar(scan_sequence1, SPCM0_1click_counts_iAll, yerr=np.sqrt(SPCM0_1click_counts_iAll), ls='none', color='tab:red')
axs[0].errorbar(scan_sequence1, SPCM0_2click_counts_iAll, yerr=np.sqrt(SPCM0_2click_counts_iAll), ls='none', color='tab:blue')
axs[0].set_title('SPCM0')
axs[0].set_ylabel('Occurrences')
axs[0].set_xlabel(scan_variable1_name);

### SPCM1
axs[1].plot(scan_sequence1, SPCM1_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[1].plot(scan_sequence1, SPCM1_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[1].errorbar(scan_sequence1, SPCM1_1click_counts_iAll, yerr=np.sqrt(SPCM1_1click_counts_iAll), ls='none', color='tab:red')
axs[1].errorbar(scan_sequence1, SPCM1_2click_counts_iAll, yerr=np.sqrt(SPCM1_2click_counts_iAll), ls='none', color='tab:blue')
axs[1].set_title('SPCM1')
axs[1].set_ylabel('Occurrences')
axs[1].set_xlabel(scan_variable1_name);

### SPCM0_OtherNode
axs[2].plot(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[2].plot(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[2].set_title('SPCM0_OtherNode')
axs[2].set_ylabel('Occurrences')
axs[2].set_xlabel(scan_variable1_name);

### SPCM1_OtherNode
axs[3].plot(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[3].plot(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[3].set_title('SPCM1_OtherNode')
axs[3].set_ylabel('Occurrences')
axs[3].set_xlabel(scan_variable1_name);

### all SPCMs
axs[4].plot(scan_sequence1, n_excitation_cycles_iAll * Col_Eff, label='Max expected', ls='solid', color='tab:gray', marker='o') ### I assumed only one excitation can happen with n_excitation_attempts
axs[4].plot(scan_sequence1, AllSPCMs_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[4].plot(scan_sequence1, AllSPCMs_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[4].errorbar(scan_sequence1, AllSPCMs_1click_counts_iAll, yerr=np.sqrt(AllSPCMs_1click_counts_iAll), ls='none', color='tab:red')
axs[4].errorbar(scan_sequence1, AllSPCMs_2click_counts_iAll, yerr=np.sqrt(AllSPCMs_2click_counts_iAll), ls='none', color='tab:blue')
axs[4].set_title('All SPCMs')
axs[4].set_ylabel('Occurrences')
axs[4].set_xlabel(scan_variable1_name);
axs[4].legend(loc=(1.05,0.2))
##########################################################################################




###################################################### plot of 2-click counts
fig, axs = plt.subplots(1, 5, figsize=(18, 3), sharey=True)

### SPCM0
# axs[0].plot(scan_sequence1, SPCM0_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[0].plot(scan_sequence1, SPCM0_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[0].errorbar(scan_sequence1, SPCM0_1click_counts_iAll, yerr=np.sqrt(SPCM0_1click_counts_iAll), ls='none', color='tab:red')
axs[0].errorbar(scan_sequence1, SPCM0_2click_counts_iAll, yerr=np.sqrt(SPCM0_2click_counts_iAll), ls='none', color='tab:blue')
axs[0].set_title('SPCM0')
axs[0].set_ylabel('Occurrences')
axs[0].set_xlabel(scan_variable1_name);

### SPCM1
# axs[1].plot(scan_sequence1, SPCM1_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[1].plot(scan_sequence1, SPCM1_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[1].errorbar(scan_sequence1, SPCM1_1click_counts_iAll, yerr=np.sqrt(SPCM1_1click_counts_iAll), ls='none', color='tab:red')
axs[1].errorbar(scan_sequence1, SPCM1_2click_counts_iAll, yerr=np.sqrt(SPCM1_2click_counts_iAll), ls='none', color='tab:blue')
axs[1].set_title('SPCM1')
axs[1].set_ylabel('Occurrences')
axs[1].set_xlabel(scan_variable1_name);

### SPCM0_OtherNode
# axs[2].plot(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[2].plot(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[2].set_title('SPCM0_OtherNode')
axs[2].set_ylabel('Occurrences')
axs[2].set_xlabel(scan_variable1_name);

### SPCM1_OtherNode
# axs[3].plot(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[3].plot(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[3].set_title('SPCM1_OtherNode')
axs[3].set_ylabel('Occurrences')
axs[3].set_xlabel(scan_variable1_name);

### all SPCMs
# axs[4].plot(scan_sequence1, n_excitation_cycles_iAll * Col_Eff, label='Max expected', ls='solid', color='tab:gray', marker='o') ### I assumed only one excitation can happen with n_excitation_attempts
# axs[4].plot(scan_sequence1, AllSPCMs_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[4].plot(scan_sequence1, AllSPCMs_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[4].errorbar(scan_sequence1, AllSPCMs_1click_counts_iAll, yerr=np.sqrt(AllSPCMs_1click_counts_iAll), ls='none', color='tab:red')
axs[4].errorbar(scan_sequence1, AllSPCMs_2click_counts_iAll, yerr=np.sqrt(AllSPCMs_2click_counts_iAll), ls='none', color='tab:blue')
axs[4].set_title('All SPCMs')
axs[4].set_ylabel('Occurrences')
axs[4].set_xlabel(scan_variable1_name);
axs[4].legend(loc=(1.05,0.2))
##########################################################################################


AllSPCMs_rel_tStamps_iAll = [
    np.concatenate((spcm0, spcm1, spcm0_other, spcm1_other))
    for spcm0, spcm1, spcm0_other, spcm1_other in zip(
        SPCM0_rel_tStamps_iAll,
        SPCM1_rel_tStamps_iAll,
        SPCM0_OtherNode_rel_tStamps_iAll,
        SPCM1_OtherNode_rel_tStamps_iAll,
    )
]

###################################################### plot of photon histograms
if show_photon_hist:
    for i in range(iterations):
        fig, axs = plt.subplots(1, 5, figsize=(18, 3), sharey=True, sharex=True)
    
        axs[0].hist(SPCM0_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[0].set_xlabel("detection time (ns)")
        axs[0].set_ylabel("Occurence")
        axs[0].set_title("SPCM0 photons")
    
        axs[1].hist(SPCM1_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[1].set_xlabel("detection time (ns)")
        axs[1].set_title("SPCM1 photons");
    
        axs[2].hist(SPCM0_OtherNode_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[2].set_xlabel("detection time (ns)")
        axs[2].set_title("SPCM0_OtherNode photons")
    
        axs[3].hist(SPCM1_OtherNode_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[3].set_xlabel("detection time (ns)")
        axs[3].set_title("SPCM1_OtherNode photons");
    
        axs[4].hist(AllSPCMs_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7, label=f'{scan_variable1_name}: {scan_sequence1[i]}')
        axs[4].set_xlabel("detection time (ns)")
        axs[4].set_title("All SPCMs photons")
        axs[4].legend(loc=(1.05, 0.2))
##########################################################################################


fig, axs = plt.subplots(1, 3, figsize=(12, 3))
########################################## Plot average number of excitation_cycles per measuremnt
axs[0].plot(scan_sequence1, n_excitation_cycles_iAll/n_measurements, label='ave excitation cycle', marker='o', color='tab:gray')
axs[0].set_xlabel(scan_variable1_name)
axs[0].set_ylabel('n_excitation_cycles')
axs[0].set_title('average n. of excitation cycles')

########################################## Plot g2
axs[1].plot(scan_sequence1, g2_0, marker='o', color='tab:red')
axs[1].errorbar(scan_sequence1, g2_0, yerr=g2_0_err, ls='none',color='tab:red')
axs[1].set_xlabel(scan_variable1_name)
# axs[1].set_ylabel('g2(0)')
axs[1].set_title('g2(0)');

########################################## Plot avergae rate
axs[2].plot(scan_sequence1, Ave_photon_rate_iAll, marker='o', color='tab:red')
# axs[2].legend(loc=(1.05,0.5))
axs[2].set_xlabel(scan_variable1_name)
axs[2].set_ylabel('rate (1/s)')
axs[2].set_title('average rate');

plt.subplots_adjust(wspace=0.4)  # Increase spacing between subplots
plt.show()
```

## combining files

### helper functions

```python
def get_full_path_from_index(file_index):
    f = fnames[file_index]
    return f if os.path.isabs(f) else os.path.join(results, f)


def read_h5_datasets(file_index):
    fname = get_full_path_from_index(file_index)

    data = {}
    with h5py.File(fname, "r") as h5:
        ds = h5["datasets"]
        for key in ds.keys():
            x = ds[key][()]
            if isinstance(x, bytes):
                x = x.decode()
            data[key] = x

    return data


def scalar_if_possible(x):
    x = np.asarray(x)
    if x.shape == ():
        return x.item()
    if x.size == 1:
        return x.ravel()[0].item()
    return x


def get_n_excitation_cycles_iAll(n_excitation_cycles, iterations, n_measurements):
    """
    Same logic as the working individual-file analysis:
        n_excitation_cycles_iAll[i] = sum(n_excitation_cycles[i*n_measurements+1:(i+1)*n_measurements+1] + 1)
    """
    n_excitation_cycles = np.asarray(n_excitation_cycles).ravel()
    n_excitation_cycles_iAll = np.zeros(iterations, dtype=int)

    for i in range(iterations):
        n_excitation_cycles_iAll[i] = np.sum(
            np.asarray(
                n_excitation_cycles[i*n_measurements+1:(i+1)*n_measurements+1],
                dtype=int,
            ) + 1
        )

    return n_excitation_cycles_iAll


def get_n_measurements_for_file(data, iterations):
    if "measurements" in data:
        return int(scalar_if_possible(data["measurements"]))

    if "n_measurements" in data:
        return int(scalar_if_possible(data["n_measurements"]))

    if "n_excitation_cycles" in data:
        return len(np.asarray(data["n_excitation_cycles"]).ravel()) // iterations

    raise KeyError("No measurements / n_measurements dataset found in this file.")


def get_n_excitation_cycles_iAll_from_data(data, iterations=None):
    """
    Return excitation cycles per scan point for THIS file.
    This avoids accidentally using a global n_excitation_cycles_iAll from another file.
    """
    if iterations is None:
        scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
        iterations = len(scan_sequence1)

    if "n_excitation_cycles_iAll" in data:
        return np.asarray(data["n_excitation_cycles_iAll"], dtype=int).ravel()

    if "n_excitation_cycles" in data:
        n_excitation_cycles = np.asarray(data["n_excitation_cycles"]).ravel()
        n_measurements_file = get_n_measurements_for_file(data, iterations)

        if n_excitation_cycles.size == 1:
            # Fallback for files that store only one constant value.
            return np.full(
                iterations,
                int(scalar_if_possible(n_excitation_cycles)) * n_measurements_file,
                dtype=int,
            )

        return get_n_excitation_cycles_iAll(
            n_excitation_cycles,
            iterations,
            n_measurements_file,
        )

    raise KeyError("No excitation-cycle dataset found in this file.")


def get_start_end_indices_for_file(data, n_excitation_cycles_iAll_file=None):
    """
    IMPORTANT FIX.

    Do NOT split the timestamp arrays into equal-sized chunks.
    Individual-file analysis uses:
        n_tStamps_iAll = n_excitation_cycles_iAll * n_excitation_attempts
        end_indices = np.cumsum(n_tStamps_iAll)
        start_indices = np.insert(end_indices[:-1], 0, 0)

    The accumulated-file analysis has to use the same boundaries.
    Otherwise photon timestamps from one scan point can be subtracted from
    reference timestamps belonging to a neighboring scan point, which creates
    fake out-of-window arrival times.
    """
    scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
    iterations = len(scan_sequence1)

    if n_excitation_cycles_iAll_file is None:
        n_excitation_cycles_iAll_file = get_n_excitation_cycles_iAll_from_data(
            data,
            iterations,
        )

    if "n_excitation_attempts" in data:
        n_excitation_attempts_file = int(scalar_if_possible(data["n_excitation_attempts"]))
    else:
        n_excitation_attempts_file = 1

    n_tStamps_iAll_file = np.asarray(n_excitation_cycles_iAll_file, dtype=int) * n_excitation_attempts_file

    end_indices = np.cumsum(n_tStamps_iAll_file)
    start_indices = np.insert(end_indices[:-1], 0, 0)

    n_total_reference = len(data["reference_tStamps_t1"]) - 1
    if int(end_indices[-1]) != int(n_total_reference):
        print(
            "WARNING: timestamp-boundary mismatch for this file. "
            f"sum(n_tStamps_iAll) = {int(end_indices[-1])}, "
            f"len(reference_tStamps_t1)-1 = {int(n_total_reference)}. "
            "Check n_excitation_cycles_iAll / n_excitation_cycles / n_excitation_attempts."
        )

    return start_indices, end_indices


def rel_timestamps_ns(tStamps_i, t1_reference_i):
    """
    Convert timestamp array for one scan point into relative detection times in ns.

    IMPORTANT:
    This now matches the individual-file analysis logic in get_rel_tStamps_i():
      - remove -1 no-click entries
      - subtract the matching t1 reference for that same shot
      - flatten 1-click and 2-click timestamps
      - remove negative relative timestamps

    The previous accumulated-file helper kept negative relative timestamps.
    That made the accumulated histogram look different from the individual-file histogram.
    """
    tStamps_i = np.asarray(tStamps_i)
    t1_reference_i = np.asarray(t1_reference_i)

    if tStamps_i.size == 0 or t1_reference_i.size == 0:
        return np.array([], dtype=float)

    # Make the shape shot x click_index.
    # For the normal max_clicks=2 case this is already 2D.
    if tStamps_i.ndim == 1:
        tStamps_i = tStamps_i.reshape(-1, 1)

    # Match the original individual-file logic, which uses zip(...)
    # and therefore only compares rows with existing references.
    n = min(len(tStamps_i), len(t1_reference_i))
    tStamps_i = tStamps_i[:n]
    t1_reference_i = t1_reference_i[:n]

    rel_chunks = [
        row[row != -1] - ref
        for row, ref in zip(tStamps_i, t1_reference_i)
    ]

    if len(rel_chunks) == 0:
        return np.array([], dtype=float)

    rel_tStamps_i = np.concatenate(rel_chunks) * 1e9
    rel_tStamps_i = rel_tStamps_i[np.isfinite(rel_tStamps_i)]

    # Same filter as the individual-file analysis:
    # in rare cases a photon timestamp is earlier than its reference.
    # Those were removed in the working individual-file path.
    rel_tStamps_i = rel_tStamps_i[rel_tStamps_i >= 0]

    return rel_tStamps_i


def first_click_exists(tStamps_i):
    tStamps_i = np.asarray(tStamps_i)

    if tStamps_i.ndim == 1:
        return tStamps_i != -1

    return tStamps_i[:, 0] != -1


def analyze_one_file(file_index):
    data = read_h5_datasets(file_index)

    scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
    iterations = len(scan_sequence1)

    # Get excitation cycles first, then use them to compute the SAME
    # start/end indices as the working individual-file analysis.
    n_excitation_cycles_iAll_file = get_n_excitation_cycles_iAll_from_data(data, iterations)
    start_indices, end_indices = get_start_end_indices_for_file(data, n_excitation_cycles_iAll_file)

    if "Atom_loading_time" in data:
        all_atom_loading_times = [np.asarray(data["Atom_loading_time"]).ravel()[1:]]
    else:
        all_atom_loading_times = []

    SPCM0_rel_tStamps_iAll = []
    SPCM1_rel_tStamps_iAll = []
    SPCM0_OtherNode_rel_tStamps_iAll = []
    SPCM1_OtherNode_rel_tStamps_iAll = []
    AllSPCMs_rel_tStamps_iAll = []

    H1H2_clicks_iAll = np.array([])
    V1V2_clicks_iAll = np.array([])
    H1V2_clicks_iAll = np.array([])
    V1H2_clicks_iAll = np.array([])

    H1V1_clicks_iAll = np.array([])
    H2V2_clicks_iAll = np.array([])

    H1orV1_and_H2orV2_clicks_iAll = np.array([])

    for i in range(iterations):
        start = start_indices[i]
        end = end_indices[i]

        SPCM0_tStamps_i = get_tStamps_i(
            data["SPCM0_Photon_tStamps"],
            start,
            end,
        )  # H1

        SPCM1_tStamps_i = get_tStamps_i(
            data["SPCM1_Photon_tStamps"],
            start,
            end,
        )  # V1

        SPCM0_OtherNode_tStamps_i = get_tStamps_i(
            data["SPCM0_OtherNode_Photon_tStamps"],
            start,
            end,
        )  # H2

        SPCM1_OtherNode_tStamps_i = get_tStamps_i(
            data["SPCM1_OtherNode_Photon_tStamps"],
            start,
            end,
        )  # V2

        t1_reference_i = data["reference_tStamps_t1"][1:][start:end]

        # Compute the relative timestamp arrays detector-by-detector.
        # Then build AllSPCMs_rel_tStamps_iAll from these four arrays.
        SPCM0_rel_i = rel_timestamps_ns(SPCM0_tStamps_i, t1_reference_i)
        SPCM1_rel_i = rel_timestamps_ns(SPCM1_tStamps_i, t1_reference_i)
        SPCM0_OtherNode_rel_i = rel_timestamps_ns(SPCM0_OtherNode_tStamps_i, t1_reference_i)
        SPCM1_OtherNode_rel_i = rel_timestamps_ns(SPCM1_OtherNode_tStamps_i, t1_reference_i)

        SPCM0_rel_tStamps_iAll.append(SPCM0_rel_i)
        SPCM1_rel_tStamps_iAll.append(SPCM1_rel_i)
        SPCM0_OtherNode_rel_tStamps_iAll.append(SPCM0_OtherNode_rel_i)
        SPCM1_OtherNode_rel_tStamps_iAll.append(SPCM1_OtherNode_rel_i)

        AllSPCMs_rel_tStamps_iAll.append(
            np.concatenate((
                SPCM0_rel_i,
                SPCM1_rel_i,
                SPCM0_OtherNode_rel_i,
                SPCM1_OtherNode_rel_i,
            ))
        )

        H1 = first_click_exists(SPCM0_tStamps_i)
        V1 = first_click_exists(SPCM1_tStamps_i)
        H2 = first_click_exists(SPCM0_OtherNode_tStamps_i)
        V2 = first_click_exists(SPCM1_OtherNode_tStamps_i)

        n = min(len(H1), len(V1), len(H2), len(V2))
        H1, V1, H2, V2 = H1[:n], V1[:n], H2[:n], V2[:n]

        H1H2_clicks_iAll = np.append(H1H2_clicks_iAll, np.sum(H1 & H2))
        V1V2_clicks_iAll = np.append(V1V2_clicks_iAll, np.sum(V1 & V2))
        H1V2_clicks_iAll = np.append(H1V2_clicks_iAll, np.sum(H1 & V2))
        V1H2_clicks_iAll = np.append(V1H2_clicks_iAll, np.sum(V1 & H2))

        H1V1_clicks_iAll = np.append(H1V1_clicks_iAll, np.sum(H1 & V1))
        H2V2_clicks_iAll = np.append(H2V2_clicks_iAll, np.sum(H2 & V2))

        H1orV1_and_H2orV2_clicks_iAll = np.append(
            H1orV1_and_H2orV2_clicks_iAll,
            np.sum((H1 | V1) & (H2 | V2)),
        )

    return {
        "file_index": file_index,
        "scan_sequence1": scan_sequence1,
        "scan_variable1_name": data["scan_variable1_name"],

        "n_excitation_cycles_iAll": n_excitation_cycles_iAll_file,

        "SPCM0_rel_tStamps_iAll": SPCM0_rel_tStamps_iAll,
        "SPCM1_rel_tStamps_iAll": SPCM1_rel_tStamps_iAll,
        "SPCM0_OtherNode_rel_tStamps_iAll": SPCM0_OtherNode_rel_tStamps_iAll,
        "SPCM1_OtherNode_rel_tStamps_iAll": SPCM1_OtherNode_rel_tStamps_iAll,
        "AllSPCMs_rel_tStamps_iAll": AllSPCMs_rel_tStamps_iAll,

        "H1H2": H1H2_clicks_iAll,
        "V1V2": V1V2_clicks_iAll,
        "H1V2": H1V2_clicks_iAll,
        "V1H2": V1H2_clicks_iAll,

        "H1V1": H1V1_clicks_iAll,
        "H2V2": H2V2_clicks_iAll,

        "H1orV1_and_H2orV2": H1orV1_and_H2orV2_clicks_iAll,

        "all_atom_loading_times": all_atom_loading_times,
    }


import re
from IPython.display import display, HTML


def get_rid_from_file(file_index):
    fname = get_full_path_from_index(file_index)

    rid = None

    with h5py.File(fname, "r") as h5:
        # Try h5 attributes first
        for key in ["rid", "RID", "run_id"]:
            if key in h5.attrs:
                rid = h5.attrs[key]
                break

        # Try datasets group
        if rid is None and "datasets" in h5:
            ds = h5["datasets"]
            for key in ["rid", "RID", "run_id"]:
                if key in ds:
                    rid = ds[key][()]
                    break

    if isinstance(rid, bytes):
        rid = rid.decode()

    if rid is not None:
        return scalar_if_possible(rid)

    # Fallback: try to extract RID from filename
    base = os.path.basename(fname)
    match = re.search(r"(?:RID|rid)[_\-\s]*(\d+)", base)

    if match is None:
        match = re.search(r"(\d{4,})", base)

    return match.group(1) if match else "unknown"


def get_file_summary_row(file_index):
    data = read_h5_datasets(file_index)

    scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
    iterations = len(scan_sequence1)

    if "measurements" in data:
        measurements = scalar_if_possible(data["measurements"])
    elif "n_measurements" in data:
        measurements = scalar_if_possible(data["n_measurements"])
    else:
        measurements = "unknown"

    try:
        n_excitation_cycles_iAll_file = get_n_excitation_cycles_iAll_from_data(data, iterations)
        total_excitation_cycles = int(np.sum(n_excitation_cycles_iAll_file))
    except Exception:
        total_excitation_cycles = len(data["reference_tStamps_t1"]) - 1

    return {
        "file_index": file_index,
        "RID": get_rid_from_file(file_index),
        "# measurements": measurements,
        "# total excitation cycles": total_excitation_cycles,
    }


def display_selected_file_summary(selected_file_indices):
    rows = [get_file_summary_row(file_index) for file_index in selected_file_indices]

    numeric_measurements = [row["# measurements"] for row in rows if isinstance(row["# measurements"], (int, float, np.integer, np.floating))]
    total_measurements = np.sum(numeric_measurements) if len(numeric_measurements) else "unknown"
    total_excitation_cycles = np.sum([row["# total excitation cycles"] for row in rows])

    html = """
    <table>
        <tr>
            <th>file_index</th>
            <th>RID</th>
            <th># measurements</th>
            <th># total excitation cycles</th>
        </tr>
    """

    for row in rows:
        html += f"""
        <tr>
            <td>{row["file_index"]}</td>
            <td>{row["RID"]}</td>
            <td>{row["# measurements"]}</td>
            <td>{row["# total excitation cycles"]}</td>
        </tr>
        """

    html += f"""
        <tr>
            <td><b>TOTAL</b></td>
            <td></td>
            <td><b>{total_measurements}</b></td>
            <td><b>{total_excitation_cycles}</b></td>
        </tr>
    """

    html += "</table>"

    display(HTML(html))

    return rows
```

### select files and combine

```python
# ============================================================
# Select files to accumulate
# ============================================================

selected_file_indices = [
    # 123, 136, 137, 148, 161, 162, 163,  #### ~2026-06-27 - node1 rate not good; some synchronization issue;
    201, 202, 204, 205, #### 2026-06-29 - node1 rate fixed; synchronization fixed;
    206, 209, 210, 211,
    212, 213, 214, 215, 
    216, 217, 220, 221,

]

selected_file_summary = display_selected_file_summary(selected_file_indices)

# ============================================================
# Analyze selected files
# ============================================================

file_results = [analyze_one_file(file_index) for file_index in selected_file_indices]

scan_sequence1 = file_results[0]["scan_sequence1"]
scan_variable1_name = file_results[0]["scan_variable1_name"]
iterations = len(scan_sequence1)

# Check that all selected files have the same scan sequence
for r in file_results[1:]:
    if len(r["scan_sequence1"]) != len(scan_sequence1) or not np.allclose(r["scan_sequence1"], scan_sequence1):
        raise ValueError(
            f"File index {r['file_index']} does not have the same scan_sequence1."
        )

# ============================================================
# Accumulate detector-specific relative timestamps first
# ============================================================

SPCM0_rel_tStamps_iAll = [
    np.concatenate([r["SPCM0_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

SPCM1_rel_tStamps_iAll = [
    np.concatenate([r["SPCM1_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

SPCM0_OtherNode_rel_tStamps_iAll = [
    np.concatenate([r["SPCM0_OtherNode_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

SPCM1_OtherNode_rel_tStamps_iAll = [
    np.concatenate([r["SPCM1_OtherNode_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

# Now build AllSPCMs_rel_tStamps_iAll in exactly the same way as the
# individual-file analysis.
AllSPCMs_rel_tStamps_iAll = [
    np.concatenate((spcm0, spcm1, spcm0_other, spcm1_other))
    for spcm0, spcm1, spcm0_other, spcm1_other in zip(
        SPCM0_rel_tStamps_iAll,
        SPCM1_rel_tStamps_iAll,
        SPCM0_OtherNode_rel_tStamps_iAll,
        SPCM1_OtherNode_rel_tStamps_iAll,
    )
]

# Diagnostic: this should now reflect only true relative timestamps.
for i, arr in enumerate(AllSPCMs_rel_tStamps_iAll):
    if len(arr) > 0:
        print(
            f"scan i={i}, scan value={scan_sequence1[i]}: "
            f"N={len(arr)}, min={np.min(arr):.3f} ns, max={np.max(arr):.3f} ns"
        )
    else:
        print(f"scan i={i}, scan value={scan_sequence1[i]}: N=0")

# ============================================================
# Accumulate coincidence counts
# ============================================================

H1H2_clicks_iAll = np.sum([r["H1H2"] for r in file_results], axis=0)
V1V2_clicks_iAll = np.sum([r["V1V2"] for r in file_results], axis=0)
H1V2_clicks_iAll = np.sum([r["H1V2"] for r in file_results], axis=0)
V1H2_clicks_iAll = np.sum([r["V1H2"] for r in file_results], axis=0)

H1V1_clicks_iAll = np.sum([r["H1V1"] for r in file_results], axis=0)
H2V2_clicks_iAll = np.sum([r["H2V2"] for r in file_results], axis=0)

H1orV1_and_H2orV2_clicks_iAll = np.sum(
    [r["H1orV1_and_H2orV2"] for r in file_results],
    axis=0,
)

H1H2_or_V1V2_clicks_iAll = H1H2_clicks_iAll + V1V2_clicks_iAll
H1V2_or_V1H2_clicks_iAll = H1V2_clicks_iAll + V1H2_clicks_iAll
H1V1_or_H2V2_clicks_iAll = H1V1_clicks_iAll + H2V2_clicks_iAll

# ============================================================
# Accumulate excitation cycles
# ============================================================

n_excitation_cycles_iAll_combined = np.sum(
    [r["n_excitation_cycles_iAll"] for r in file_results],
    axis=0,
)

# Keep this name also, because later plotting cells may expect it.
n_excitation_cycles_iAll = n_excitation_cycles_iAll_combined

# ============================================================
# Accumulate atom loading times, if present
# ============================================================

all_atom_loading_times = []
for r in file_results:
    all_atom_loading_times.extend(r.get("all_atom_loading_times", []))

print("Accumulated file indices:", selected_file_indices)
print("scan_sequence1:", scan_sequence1)
print("n_excitation_cycles_iAll_combined:", n_excitation_cycles_iAll_combined)
```

### Atom loading time analysis

```python
import math
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Plot Atom_loading_time values for selected files
# Max 7 plots per row
# ============================================================

plots_per_row = 5
n_files = len(selected_file_indices)

if n_files == 0:
    raise ValueError("selected_file_indices is empty.")

ncols = min(plots_per_row, n_files)
nrows = math.ceil(n_files / plots_per_row)

fig, axs = plt.subplots(
    nrows=nrows,
    ncols=ncols,
    figsize=(4 * ncols, 3.2 * nrows),
    squeeze=False
)

axs = axs.ravel()

for plot_i, file_index in enumerate(selected_file_indices):
    ax = axs[plot_i]

    # Analyze this file only
    result = analyze_one_file(file_index)

    all_atom_loading_times = result["all_atom_loading_times"]

    # ============================================================
    # Flatten atom loading times safely
    # This matches your working logic:
    # combined_atom_loading_times = [t for loading_times in all_atom_loading_times for t in loading_times]
    # ============================================================

    if len(all_atom_loading_times) == 0:
        combined_atom_loading_times = np.array([])
    else:
        first_item = all_atom_loading_times[0]

        # Case 1: all_atom_loading_times is a list of arrays
        if np.ndim(first_item) > 0:
            combined_atom_loading_times = np.concatenate([
                np.asarray(loading_times).ravel()
                for loading_times in all_atom_loading_times
            ])

        # Case 2: all_atom_loading_times is already one flat array/list
        else:
            combined_atom_loading_times = np.asarray(all_atom_loading_times).ravel()

    if len(combined_atom_loading_times) == 0:
        ax.set_title(f"File {file_index}: no loading times")
        ax.axis("off")
        continue

    mean_loading_time = np.mean(combined_atom_loading_times)

    ax.plot(combined_atom_loading_times, 'r-', label='Atom loading time')

    ax.set_xlabel("Atom number")
    ax.set_ylabel("Loading time (s)")
    ax.set_title(
        f"File {file_index}\nMean: {mean_loading_time:.3f} s"
    )
    ax.legend(loc='best')

    print(
        f"file {file_index}:",
        "N =", len(combined_atom_loading_times),
        "first 5 =", combined_atom_loading_times[:5],
        "mean =", mean_loading_time
    )

# Hide unused empty subplots
for j in range(n_files, len(axs)):
    axs[j].axis("off")

plt.tight_layout()
plt.show()
```

### HOM fitting

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit


# ============================================================
# Scaling counts to common excitation cycles
# ============================================================

def scale_to_common_cycles(counts, n_cycles, N_common=None):
    counts = np.asarray(counts, dtype=float)
    n_cycles = np.asarray(n_cycles, dtype=float)

    if N_common is None:
        N_common = np.mean(n_cycles)

    y_scaled = counts / n_cycles * N_common

    # Poisson error on raw counts, then scaled the same way.
    # Use max(counts, 1) so zero-count points do not get zero error.
    yerr_scaled = np.sqrt(np.maximum(counts, 1.0)) / n_cycles * N_common

    return y_scaled, yerr_scaled, N_common


# ============================================================
# Gaussian HOM dip model
# ============================================================

def hom_gaussian_dip(x, y_far, depth, x0, sigma):
    """
    Gaussian HOM dip:

        y(x) = y_far - depth * exp[-(x - x0)^2 / (2 sigma^2)]

    y_far : coincidence level far away from temporal overlap
    depth : dip depth
    x0    : dip center
    sigma : dip width
    """
    return y_far - depth * np.exp(-0.5 * ((x - x0) / sigma) ** 2)


def make_hom_initial_guess(x, y):
    """
    Automatically choose reasonable initial guesses from the data.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    good = np.isfinite(x) & np.isfinite(y)
    x = x[good]
    y = y[good]

    y_min = np.min(y)
    y_max = np.max(y)

    # Estimate far-away level from the high-count points.
    # This is more stable than using only max(y).
    n_top = max(2, int(np.ceil(0.3 * len(y))))
    y_sorted = np.sort(y)
    y_far_guess = np.mean(y_sorted[-n_top:])

    # Dip center guess: location of minimum count.
    x0_guess = x[np.argmin(y)]

    # Dip depth guess.
    depth_guess = max(y_far_guess - y_min, 0.1 * max(y_far_guess, 1.0))

    # Width guess: use scan range / 5.
    x_range = np.max(x) - np.min(x)
    sigma_guess = max(x_range / 5.0, 1.0)

    return y_far_guess, depth_guess, x0_guess, sigma_guess


def fit_single_hom_dip(
    x,
    counts,
    n_cycles,
    label,
    N_common=None,
    x_label="Timing difference in ns",
    plot_residuals=True,
):
    """
    Scale one coincidence-count dataset, fit a Gaussian HOM dip,
    and plot the data separately with the fit.
    """

    x = np.asarray(x, dtype=float)
    counts = np.asarray(counts, dtype=float)
    n_cycles = np.asarray(n_cycles, dtype=float)

    y_scaled, yerr_scaled, N_common_used = scale_to_common_cycles(
        counts,
        n_cycles,
        N_common=N_common,
    )

    # Remove bad points
    good = (
        np.isfinite(x)
        & np.isfinite(y_scaled)
        & np.isfinite(yerr_scaled)
        & (n_cycles > 0)
    )

    x_fitdata = x[good]
    y_fitdata = y_scaled[good]
    yerr_fitdata = yerr_scaled[good]

    # Sort by x
    order = np.argsort(x_fitdata)
    x_fitdata = x_fitdata[order]
    y_fitdata = y_fitdata[order]
    yerr_fitdata = yerr_fitdata[order]

    if len(x_fitdata) < 4:
        raise ValueError(f"{label}: Need at least 4 valid data points for this fit.")

    # ============================================================
    # Adaptive initial guesses
    # ============================================================
    p0 = make_hom_initial_guess(x_fitdata, y_fitdata)

    y_far_guess, depth_guess, x0_guess, sigma_guess = p0

    print("\nInitial guesses for", label)
    print("----------------------------------------")
    print(f"y_far guess = {y_far_guess:.4g}")
    print(f"depth guess = {depth_guess:.4g}")
    print(f"x0 guess    = {x0_guess:.4g} ns")
    print(f"sigma guess = {sigma_guess:.4g} ns")

    # ============================================================
    # Bounds
    # ============================================================
    y_max = np.max(y_fitdata)
    x_min = np.min(x_fitdata)
    x_max = np.max(x_fitdata)
    x_range = x_max - x_min

    lower_bounds = [
        0.0,          # y_far
        0.0,          # depth
        x_min,        # x0
        0.1,          # sigma
    ]

    upper_bounds = [
        5.0 * max(y_max, 1.0),       # y_far
        5.0 * max(y_max, 1.0),       # depth
        x_max,                       # x0
        5.0 * max(x_range, 1.0),     # sigma
    ]

    # ============================================================
    # Fit
    # ============================================================
    popt, pcov = curve_fit(
        hom_gaussian_dip,
        x_fitdata,
        y_fitdata,
        p0=p0,
        sigma=yerr_fitdata,
        absolute_sigma=True,
        bounds=(lower_bounds, upper_bounds),
        maxfev=50000,
    )

    perr = np.sqrt(np.diag(pcov))

    y_far, depth, x0, sigma = popt
    y_far_err, depth_err, x0_err, sigma_err = perr

    # HOM visibility from Gaussian dip fit
    visibility = depth / y_far if y_far != 0 else np.nan

    if depth > 0 and y_far > 0:
        visibility_err = visibility * np.sqrt(
            (depth_err / depth) ** 2 + (y_far_err / y_far) ** 2
        )
    else:
        visibility_err = np.nan

    fwhm = 2.0 * np.sqrt(2.0 * np.log(2.0)) * sigma
    fwhm_err = 2.0 * np.sqrt(2.0 * np.log(2.0)) * sigma_err

    # Fit quality
    y_model_data = hom_gaussian_dip(x_fitdata, *popt)
    residuals = y_fitdata - y_model_data

    chi2 = np.sum((residuals / yerr_fitdata) ** 2)
    dof = len(x_fitdata) - len(popt)
    reduced_chi2 = chi2 / dof if dof > 0 else np.nan

    # Smooth fit curve
    x_smooth = np.linspace(x_min, x_max, 1000)
    y_smooth = hom_gaussian_dip(x_smooth, *popt)

    # ============================================================
    # Print fit result
    # ============================================================
    print("\nFit result for", label)
    print("============================================================")
    print(f"y_far      = {y_far:.4g} ± {y_far_err:.4g}")
    print(f"depth      = {depth:.4g} ± {depth_err:.4g}")
    print(f"x0         = {x0:.4g} ± {x0_err:.4g} ns")
    print(f"sigma      = {sigma:.4g} ± {sigma_err:.4g} ns")
    print(f"FWHM       = {fwhm:.4g} ± {fwhm_err:.4g} ns")
    print(f"visibility = {visibility:.4f} ± {visibility_err:.4f}")
    print(f"reduced chi^2 = {reduced_chi2:.3g}")
    print("============================================================")

    # ============================================================
    # Plot data and fit
    # ============================================================
    fig, ax = plt.subplots(1, 1, figsize=(8, 5))

    ax.errorbar(
        x_fitdata,
        y_fitdata,
        yerr=yerr_fitdata,
        fmt="o",
        capsize=3,
        label=label,
    )

    ax.plot(
        x_smooth,
        y_smooth,
        linewidth=2,
        label=(
            f"Gaussian HOM fit\n"
            f"$x_0$ = {x0:.2f} ns, "
            f"$V$ = {100 * visibility:.1f}%, "
            f"FWHM = {fwhm:.2f} ns"
        ),
    )

    ax.axvline(x0, linestyle="--", linewidth=1, label="fit center")

    ax.set_xlabel(x_label)
    ax.set_ylabel(f"equivalent counts for {N_common_used:.0f} excitation cycles")
    ax.set_title(f"HOM dip fit: {label}")
    ax.legend()
    ax.grid(alpha=0.3)

    plt.tight_layout()
    plt.show()

    # ============================================================
    # Residual plot
    # ============================================================
    if plot_residuals:
        fig, ax = plt.subplots(1, 1, figsize=(8, 3))

        ax.errorbar(
            x_fitdata,
            residuals,
            yerr=yerr_fitdata,
            fmt="o",
            capsize=3,
        )

        ax.axhline(0, linestyle="--", linewidth=1)

        ax.set_xlabel(x_label)
        ax.set_ylabel("residual")
        ax.set_title(f"Fit residuals: {label}")
        ax.grid(alpha=0.3)

        plt.tight_layout()
        plt.show()

    return {
        "label": label,
        "x": x_fitdata,
        "y_scaled": y_fitdata,
        "yerr_scaled": yerr_fitdata,
        "N_common": N_common_used,
        "popt": popt,
        "perr": perr,
        "y_far": y_far,
        "y_far_err": y_far_err,
        "depth": depth,
        "depth_err": depth_err,
        "x0": x0,
        "x0_err": x0_err,
        "sigma": sigma,
        "sigma_err": sigma_err,
        "fwhm": fwhm,
        "fwhm_err": fwhm_err,
        "visibility": visibility,
        "visibility_err": visibility_err,
        "chi2": chi2,
        "dof": dof,
        "reduced_chi2": reduced_chi2,
        "x_smooth": x_smooth,
        "y_smooth": y_smooth,
        "residuals": residuals,
    }


# ============================================================
# Run fits separately for your two datasets
# ============================================================

x_scan = np.asarray(scan_sequence1, dtype=float) - 189
N_common = np.mean(n_excitation_cycles_iAll_combined)

# hom_fit_all_HV = fit_single_hom_dip(
#     x=x_scan,
#     counts=H1orV1_and_H2orV2_clicks_iAll,
#     n_cycles=n_excitation_cycles_iAll_combined,
#     label="(H1 or V1) and (H2 or V2)",
#     N_common=N_common,
# )

hom_fit_HH_plus_VV = fit_single_hom_dip(
    x=x_scan,
    counts=H1H2_or_V1V2_clicks_iAll,
    n_cycles=n_excitation_cycles_iAll_combined,
    label="H1H2 + V1V2",
    N_common=N_common,
    plot_residuals = False
)

# plt.savefig("HOM_dip_HH_plus_VV.png", dpi=300, bbox_inches="tight", facecolor="white")
# plt.show()
```

```python
"""
What each variable represents. Example values are mostly from RID 22496, 2025-02-26. Variables ending in _iAll contain data for all interations with length equal to "iterations".

n_measurements = 2000: directly from dataset
iterations = 5: the length of scan_sequence1, directly from dataset
n_excitation_cycles = [0,  2, 20,  8,  2,  8, ...], length = 1 + n_measurements * iterations: the number of excitation cycles in each measurement
n_excitation_cycles_iAll = [16830, 16665, 16488, 16194, 16221], length = iterations: total number of excitation cycles in each iteration as an array.
n_excitation_attempts = 1: directly from dataset. Number of excitation pulses in each excitation cycle.
n_tStamps_iAll = [16830, 16665, 16488, 16194, 16221], length = iterations: total number of timestamps in each iteration as an array.
SPCM0_Photon_tStamps = [[0, 0], [-1, -1], [-1, -1], [1.32237714e+06, -1], [-1, -1], ...], length = sum(n_tStamps_iAll) = 82399, for example. Directly from dataset.
SPCM0_tStamps_i =  [[-1, -1], [-1, -1], [1.32237714e+06, -1], [-1, -1], ...], length = elements of n_tStamps_iAll, 16830 for the first iteration, for example. The timestamps in each iteration.
reference_tStamps_t1 = [0, 1322341.65973562, 1322341.65976682, 1322341.65979802, ...], length = sum(n_tStamps_iAll) = 82399, for example. Directly from dataset.
t1_reference_i = [1322341.65973562, 1322341.65976682, 1322341.65979802, ...], length = elements of n_tStamps_iAll, 16830 for the first iteration, for example. The reference times in each iteration.
SPCM0_1click_counts_iAll = [128., 122., 172., 186., 172.], length = iterations: number of 1-click events in each iteration as an array; events in the first column of SPCM0_tStamps_i for different iteration.
SPCM0_2click_counts_iAll = [0., 0., 0., 0., 0.], length = iterations: number of 2-click events in each iteration as an array; events in the second column of SPCM0_tStamps_i for different iteration.
start_indices = [0, 16830, 33495, 49983, 66177], length = iterations: to slice arrays like SPCM0_Photon_tStamps into iterations.
    Remember that SPCM0_Photon_tStamps is a single array with the timestamps for all iterations.
end_indices = [16830, 33495, 49983, 66177, 82398], length = iterations: same as start_indices
SPCM0_rel_tStamps_i = [1099.89196062, 1088.94892037, 1098.02931547, ...], length = SPCM0_1click_counts_iAll + SPCM0_2click_counts_iAll, for example length is 128 and 122 for first two iterations.
    This array contains the timestamps relative to the reference time t1 in ns unit. -1 event are excluded and 2-click events are included.
SPCM0_rel_tStamps_iAll = [[1099.89196062, 1088.94892037, 1098.02931547, ...], [..], ..], length = iterations. Arrays of arrays containing all the relative timestamps for all iteration.
AllSPCMs_tStamps_filtered_i = [[1757511.19383094, 1757533.11171639, ...], length = total number of photons in iteration i,
    i.e. length = all 1-click and 2-click counts from SPCM0, SPCM1, SPCM0_OtherNode, and SPCM1_OtherNode. Contains all the absolute time stamps from all 4 SPCMs with -1 removed and sorted from small to large.
AllSPCMs_tStamps_filtered_iAll = [[1757511.19383094, 1757533.11171639, ...], [1749316.29104084, 1749328.33163165, ..], ..], length = iterations. Array of arrays keeping all
    AllSPCMs_tStamps_filtered_i from all iterations.
Ave_photon_rate_iAll = [[0.11084697, 0.1147433 , 0.10952382, 0.1337287 , 0.12123424]], length = iterations. Contains the average photon generation rate for all iterations.
n_coincidence_iAll = [0,0,1,0,2], length = iterations. The number of coincidences in each iteration used to calculate g2.

Other notes:
1- Does this work if I change max_clicks from 2 to 1 or 3? The answer is no.

"""
show_photon_hist = False
### Print the title of the output
display(HTML(f"""<b style='font-size:18px;'> {date_filters[0]} - RID {rid} - {experiment_function}</b>"""))

Col_Eff = 0.05 # overal expected/theoretical collection and detection efficiency

### Instead of using a large n_measurements, which can sometimes lead to underflow error, we use dummy_variable to split the measurements.
### So, if scan_variable1_name == 'dummy_variable', we want to combine all iterations into one:
if scan_variable1_name == 'dummy_variable':
    n_measurements = n_measurements * iterations
    iterations = 1
    scan_sequence1 = [0]

## To look at specific iterations, you can select in scan_sequence_1, for example scan_sequence1 = [0,2]. For now, the first element has to be 0.
# scan_sequence1 = [0,4]
# iterations = len(scan_sequence1)
# n_measurements = n_measurements * iterations


n_excitation_cycles_iAll = np.zeros(iterations, dtype=int) ### total number of excitation cycle in each iteration.
n_tStamps_iAll = np.zeros(iterations, dtype=int)  ### number of timestamps in each iteration.
for i in range(iterations):
    n_excitation_cycles_iAll[i] = sum(n_excitation_cycles[i*n_measurements+1:(i+1)*n_measurements+1]+1)
    if scan_variable1_name == 'n_excitation_attempts':
        n_excitation_attempts = scan_sequence1[i]
    n_tStamps_iAll[i] = n_excitation_cycles_iAll[i] * n_excitation_attempts
    ### Note that n_excitation_cycles varies in each cycle, each contains n_excitation_attempts.
    ### the first +1 is because the first element in each dataset is 0.
    ### the last +1 is because there is one more cycle than n_excitation_cycles in each loop. See the oscciloscope. For example, if n_excitation_cycles[j] = 2,
    ### there are 2+1=3 sets of excitation cycles for that measurement.


SPCM0_1click_counts_iAll = np.array([])
SPCM0_2click_counts_iAll = np.array([])
SPCM1_1click_counts_iAll = np.array([])
SPCM1_2click_counts_iAll = np.array([])
SPCM0_OtherNode_1click_counts_iAll = np.array([])
SPCM0_OtherNode_2click_counts_iAll = np.array([])
SPCM1_OtherNode_1click_counts_iAll = np.array([])
SPCM1_OtherNode_2click_counts_iAll = np.array([])

AllSPCMs_tStamps_filtered_iAll = []

SPCM0_rel_tStamps_iAll = []
SPCM1_rel_tStamps_iAll = []
SPCM0_OtherNode_rel_tStamps_iAll = []
SPCM1_OtherNode_rel_tStamps_iAll = []

n_coincidence_iAll = np.zeros(iterations)
g2_0 = np.zeros(iterations)
g2_0_err = np.zeros(iterations)
Ave_photon_rate_iAll = np.zeros(iterations)


def get_rel_tStamps_i(tStamps_i, t1_reference_i, spcm_name):
    ### the timestamps relative to reference time. -1 events are excluded.
    rel_tStamps_i_temp = [row[row != -1] - ref for row, ref in zip(tStamps_i, t1_reference_i)]
    if not rel_tStamps_i_temp: ### np.concatenate crashes on empty lists; if no photon detected, I replace timetags by [0].
        rel_tStamps_i_temp = [[0]]
    rel_tStamps_i_temp2 = np.concatenate(rel_tStamps_i_temp) * 1e9 ## Flatten to combine 1-click and 2-click events.
    rel_tStamps_i = rel_tStamps_i_temp2[rel_tStamps_i_temp2 >= 0]
    ## in 1 out of 1million case, the reference time is later than the timetag and I get a negative
    ### element which messes up the histograms. I am simply removing these elements if any, and printing
    if len(rel_tStamps_i) != len(rel_tStamps_i_temp2):
        print(f'Early time stamp for {len(rel_tStamps_i_temp2) - len(rel_tStamps_i)} elements on {spcm_name} was detected.')
    return rel_tStamps_i


def get_tStamps_i(tStamps_all, start, end):
    tStamps_i = np.asarray(tStamps_all[1:][start:end])
    if tStamps_i.size == 0:
        return np.empty((0, 2))
    if tStamps_i.ndim == 1:
        return tStamps_i.reshape(-1, 2)
    return tStamps_i


### Calculate the ending and starting indices for the iterations
end_indices = np.cumsum(n_tStamps_iAll)
start_indices = np.insert(end_indices[:-1], 0, 0) ### prepend a 0 for the starting index

for i in range(iterations):
    start = start_indices[i]
    end = end_indices[i]
    SPCM0_tStamps_i = get_tStamps_i(SPCM0_Photon_tStamps, start, end)
    SPCM1_tStamps_i = get_tStamps_i(SPCM1_Photon_tStamps, start, end)
    SPCM0_OtherNode_tStamps_i = get_tStamps_i(SPCM0_OtherNode_Photon_tStamps, start, end)
    SPCM1_OtherNode_tStamps_i = get_tStamps_i(SPCM1_OtherNode_Photon_tStamps, start, end)
    t1_reference_i = reference_tStamps_t1[1:][start:end]

    ### number of 1-click events (the first column is none -1)
    SPCM0_1click_counts_iAll = np.append(SPCM0_1click_counts_iAll, np.count_nonzero(SPCM0_tStamps_i[:, 0] != -1))
    SPCM1_1click_counts_iAll = np.append(SPCM1_1click_counts_iAll, np.count_nonzero(SPCM1_tStamps_i[:, 0] != -1))
    SPCM0_OtherNode_1click_counts_iAll = np.append(SPCM0_OtherNode_1click_counts_iAll, np.count_nonzero(SPCM0_OtherNode_tStamps_i[:, 0] != -1))
    SPCM1_OtherNode_1click_counts_iAll = np.append(SPCM1_OtherNode_1click_counts_iAll, np.count_nonzero(SPCM1_OtherNode_tStamps_i[:, 0] != -1))

    ### number of 2-click events (the 2nd column is none -1)
    SPCM0_2click_counts_iAll = np.append(SPCM0_2click_counts_iAll, np.count_nonzero(SPCM0_tStamps_i[:, 1] != -1))
    SPCM1_2click_counts_iAll = np.append(SPCM1_2click_counts_iAll, np.count_nonzero(SPCM1_tStamps_i[:, 1] != -1))
    SPCM0_OtherNode_2click_counts_iAll = np.append(SPCM0_OtherNode_2click_counts_iAll, np.count_nonzero(SPCM0_OtherNode_tStamps_i[:, 1] != -1))
    SPCM1_OtherNode_2click_counts_iAll = np.append(SPCM1_OtherNode_2click_counts_iAll, np.count_nonzero(SPCM1_OtherNode_tStamps_i[:, 1] != -1))

    SPCM0_rel_tStamps_i = get_rel_tStamps_i(SPCM0_tStamps_i, t1_reference_i, 'SPCM0')
    SPCM1_rel_tStamps_i = get_rel_tStamps_i(SPCM1_tStamps_i, t1_reference_i, 'SPCM1')
    SPCM0_OtherNode_rel_tStamps_i = get_rel_tStamps_i(SPCM0_OtherNode_tStamps_i, t1_reference_i, 'SPCM0_OtherNode')
    SPCM1_OtherNode_rel_tStamps_i = get_rel_tStamps_i(SPCM1_OtherNode_tStamps_i, t1_reference_i, 'SPCM1_OtherNode')

    SPCM0_rel_tStamps_iAll.append(SPCM0_rel_tStamps_i)
    SPCM1_rel_tStamps_iAll.append(SPCM1_rel_tStamps_i)
    SPCM0_OtherNode_rel_tStamps_iAll.append(SPCM0_OtherNode_rel_tStamps_i)
    SPCM1_OtherNode_rel_tStamps_iAll.append(SPCM1_OtherNode_rel_tStamps_i)

    ### calculate g2 for each iteration. In future, we can also select a narrower coincidence window.
    ### For 4 SPCMs, count coincidences between all unordered detector pairs.
    first_click_arrays_i = [
        SPCM0_tStamps_i[:, 0],
        SPCM1_tStamps_i[:, 0],
        SPCM0_OtherNode_tStamps_i[:, 0],
        SPCM1_OtherNode_tStamps_i[:, 0],
    ]
    first_click_counts_i = [
        SPCM0_1click_counts_iAll[i],
        SPCM1_1click_counts_iAll[i],
        SPCM0_OtherNode_1click_counts_iAll[i],
        SPCM1_OtherNode_1click_counts_iAll[i],
    ]

    n_coincidence_i = 0
    g2_denominator_i = 0
    g2_denominator_var_i = 0
    for j in range(len(first_click_arrays_i)):
        for k in range(j + 1, len(first_click_arrays_i)):
            n_coincidence_i += np.sum((first_click_arrays_i[j] >= 1) & (first_click_arrays_i[k] >= 1))

            Nj = first_click_counts_i[j]
            Nk = first_click_counts_i[k]
            g2_denominator_i += Nj * Nk
            g2_denominator_var_i += (Nk**2) * Nj + (Nj**2) * Nk

    # n_coincidence_i = 1 ### if there is no coincidence, you can set n_coincidence_i = 1 manually to find the bound to g2 because the next count might be a coincidence.
    n_coincidence_iAll[i] = n_coincidence_i ### Having a list of coincidences for each iteration helps to learn more about the g2.

    if g2_denominator_i > 0:
        g2_0_temp = (n_coincidence_i / g2_denominator_i) * n_excitation_cycles_iAll[i]
        ### Errors for g2 results are from error propagation assuming Poissonian distribution for counts:
        if n_coincidence_i > 0:
            g2_0_err_temp = g2_0_temp * np.sqrt(1/n_coincidence_i + g2_denominator_var_i/(g2_denominator_i**2))
        else:
            g2_0_err_temp = np.nan
    else:
        g2_0_temp = np.nan
        g2_0_err_temp = np.nan
    g2_0[i] = g2_0_temp
    g2_0_err[i] = g2_0_err_temp

    ############## absolute timestamps of all the photons for rate analysis
    AllSPCMs_tStamps_flatten_i = np.concatenate((
        SPCM0_tStamps_i,
        SPCM1_tStamps_i,
        SPCM0_OtherNode_tStamps_i,
        SPCM1_OtherNode_tStamps_i,
    )).flatten() ### flatten to get the timestamps of all photons (1st and 2nd clicks).
    AllSPCMs_tStamps_filtered_i = AllSPCMs_tStamps_flatten_i[AllSPCMs_tStamps_flatten_i != -1]  ### -1 events are removed
    AllSPCMs_tStamps_filtered_i = np.sort(AllSPCMs_tStamps_filtered_i) ### sorted from small to large timestamps
    AllSPCMs_tStamps_filtered_iAll.append(AllSPCMs_tStamps_filtered_i)


    # ###########checking H1H2 ###############
    # count=0
    # for j in range(len(SPCM0_tStamps_i)):
    #     if SPCM0_tStamps_i[:, 0][j] != -1 and SPCM0_OtherNode_tStamps_i[:, 0][j] != -1:
    #         count+=1
    
    # print("H1H2 ", count)
    # ###########checking V1V2 ###############
    # count=0
    # for j in range(len(SPCM1_tStamps_i)):
    #     if SPCM1_tStamps_i[:, 0][j] != -1 and SPCM1_OtherNode_tStamps_i[:, 0][j] != -1:
    #         count+=1
    
    # print("V1V2 ",count)
    
    # ###########checking (H1orV1) (H2orV2) ###############
    # count=0
    # for j in range(len(SPCM1_tStamps_i)):
    #     if (SPCM0_tStamps_i[:, 0][j] != -1 or SPCM1_tStamps_i[:, 0][j] != -1) and (SPCM0_OtherNode_tStamps_i[:, 0][j] != -1 or SPCM1_OtherNode_tStamps_i[:, 0][j] != -1):
    #         count+=1
    
    # print("(H1orV1) (H2orV2) ",count)
    


AllSPCMs_1click_counts_iAll = (
    SPCM0_1click_counts_iAll
    + SPCM1_1click_counts_iAll
    + SPCM0_OtherNode_1click_counts_iAll
    + SPCM1_OtherNode_1click_counts_iAll
)
AllSPCMs_2click_counts_iAll = (
    SPCM0_2click_counts_iAll
    + SPCM1_2click_counts_iAll
    + SPCM0_OtherNode_2click_counts_iAll
    + SPCM1_OtherNode_2click_counts_iAll
)


print(f'n_measurements = {n_measurements}')
print(f'excitation pulse width = {t_excitation_pulse/10**-9}','ns')
print(f'SPCM window = {t_photon_collection_time/10**-9}', 'ns')
print(f'excitation AOM power (dBm) = {p_excitation}')
print(f'n_excitation_attempts = {n_excitation_attempts}')
print(f'check atom avery n excitation cycles = {atom_check_every_n}')
if n_excitation_cycles_iAll[0] > 0:
    print(f'Overal photon detection efficiency (%) = {100*AllSPCMs_1click_counts_iAll[0]/n_excitation_cycles_iAll[0]}')
else:
    print('Overal photon detection efficiency (%) = nan')



#############  Average photon generation rate calculated from last timestamp minus first timestamp.
for i in range(iterations):
    if len(AllSPCMs_tStamps_filtered_iAll[i]) >= 2:
        t_duration_i = max(AllSPCMs_tStamps_filtered_iAll[i]) - min(AllSPCMs_tStamps_filtered_iAll[i])
        if t_duration_i > 0:
            Ave_photon_rate_iAll[i] = (AllSPCMs_1click_counts_iAll[i] + AllSPCMs_2click_counts_iAll[i]) / t_duration_i
        else:
            Ave_photon_rate_iAll[i] = np.nan
    else:
        Ave_photon_rate_iAll[i] = np.nan


###################################################### plot of 1-click and 2-click counts
fig, axs = plt.subplots(1, 5, figsize=(18, 3), sharey=True)

### SPCM0
axs[0].plot(scan_sequence1, SPCM0_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[0].plot(scan_sequence1, SPCM0_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[0].errorbar(scan_sequence1, SPCM0_1click_counts_iAll, yerr=np.sqrt(SPCM0_1click_counts_iAll), ls='none', color='tab:red')
axs[0].errorbar(scan_sequence1, SPCM0_2click_counts_iAll, yerr=np.sqrt(SPCM0_2click_counts_iAll), ls='none', color='tab:blue')
axs[0].set_title('SPCM0')
axs[0].set_ylabel('Occurrences')
axs[0].set_xlabel(scan_variable1_name);

### SPCM1
axs[1].plot(scan_sequence1, SPCM1_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[1].plot(scan_sequence1, SPCM1_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[1].errorbar(scan_sequence1, SPCM1_1click_counts_iAll, yerr=np.sqrt(SPCM1_1click_counts_iAll), ls='none', color='tab:red')
axs[1].errorbar(scan_sequence1, SPCM1_2click_counts_iAll, yerr=np.sqrt(SPCM1_2click_counts_iAll), ls='none', color='tab:blue')
axs[1].set_title('SPCM1')
axs[1].set_ylabel('Occurrences')
axs[1].set_xlabel(scan_variable1_name);

### SPCM0_OtherNode
axs[2].plot(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[2].plot(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[2].set_title('SPCM0_OtherNode')
axs[2].set_ylabel('Occurrences')
axs[2].set_xlabel(scan_variable1_name);

### SPCM1_OtherNode
axs[3].plot(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[3].plot(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[3].set_title('SPCM1_OtherNode')
axs[3].set_ylabel('Occurrences')
axs[3].set_xlabel(scan_variable1_name);

### all SPCMs
axs[4].plot(scan_sequence1, n_excitation_cycles_iAll * Col_Eff, label='Max expected', ls='solid', color='tab:gray', marker='o') ### I assumed only one excitation can happen with n_excitation_attempts
axs[4].plot(scan_sequence1, AllSPCMs_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[4].plot(scan_sequence1, AllSPCMs_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
axs[4].errorbar(scan_sequence1, AllSPCMs_1click_counts_iAll, yerr=np.sqrt(AllSPCMs_1click_counts_iAll), ls='none', color='tab:red')
axs[4].errorbar(scan_sequence1, AllSPCMs_2click_counts_iAll, yerr=np.sqrt(AllSPCMs_2click_counts_iAll), ls='none', color='tab:blue')
axs[4].set_title('All SPCMs')
axs[4].set_ylabel('Occurrences')
axs[4].set_xlabel(scan_variable1_name);
axs[4].legend(loc=(1.05,0.2))
##########################################################################################




###################################################### plot of 2-click counts
fig, axs = plt.subplots(1, 5, figsize=(18, 3), sharey=True)

### SPCM0
# axs[0].plot(scan_sequence1, SPCM0_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[0].plot(scan_sequence1, SPCM0_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[0].errorbar(scan_sequence1, SPCM0_1click_counts_iAll, yerr=np.sqrt(SPCM0_1click_counts_iAll), ls='none', color='tab:red')
axs[0].errorbar(scan_sequence1, SPCM0_2click_counts_iAll, yerr=np.sqrt(SPCM0_2click_counts_iAll), ls='none', color='tab:blue')
axs[0].set_title('SPCM0')
axs[0].set_ylabel('Occurrences')
axs[0].set_xlabel(scan_variable1_name);

### SPCM1
# axs[1].plot(scan_sequence1, SPCM1_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[1].plot(scan_sequence1, SPCM1_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[1].errorbar(scan_sequence1, SPCM1_1click_counts_iAll, yerr=np.sqrt(SPCM1_1click_counts_iAll), ls='none', color='tab:red')
axs[1].errorbar(scan_sequence1, SPCM1_2click_counts_iAll, yerr=np.sqrt(SPCM1_2click_counts_iAll), ls='none', color='tab:blue')
axs[1].set_title('SPCM1')
axs[1].set_ylabel('Occurrences')
axs[1].set_xlabel(scan_variable1_name);

### SPCM0_OtherNode
# axs[2].plot(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[2].plot(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[2].errorbar(scan_sequence1, SPCM0_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM0_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[2].set_title('SPCM0_OtherNode')
axs[2].set_ylabel('Occurrences')
axs[2].set_xlabel(scan_variable1_name);

### SPCM1_OtherNode
# axs[3].plot(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[3].plot(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_1click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_1click_counts_iAll), ls='none', color='tab:red')
axs[3].errorbar(scan_sequence1, SPCM1_OtherNode_2click_counts_iAll, yerr=np.sqrt(SPCM1_OtherNode_2click_counts_iAll), ls='none', color='tab:blue')
axs[3].set_title('SPCM1_OtherNode')
axs[3].set_ylabel('Occurrences')
axs[3].set_xlabel(scan_variable1_name);

### all SPCMs
# axs[4].plot(scan_sequence1, n_excitation_cycles_iAll * Col_Eff, label='Max expected', ls='solid', color='tab:gray', marker='o') ### I assumed only one excitation can happen with n_excitation_attempts
# axs[4].plot(scan_sequence1, AllSPCMs_1click_counts_iAll, label='1-click events', ls='solid', color='tab:red', marker='o')
axs[4].plot(scan_sequence1, AllSPCMs_2click_counts_iAll, label='2-click events', ls='dashed', color='tab:blue', marker='o')
# axs[4].errorbar(scan_sequence1, AllSPCMs_1click_counts_iAll, yerr=np.sqrt(AllSPCMs_1click_counts_iAll), ls='none', color='tab:red')
axs[4].errorbar(scan_sequence1, AllSPCMs_2click_counts_iAll, yerr=np.sqrt(AllSPCMs_2click_counts_iAll), ls='none', color='tab:blue')
axs[4].set_title('All SPCMs')
axs[4].set_ylabel('Occurrences')
axs[4].set_xlabel(scan_variable1_name);
axs[4].legend(loc=(1.05,0.2))
##########################################################################################


AllSPCMs_rel_tStamps_iAll = [
    np.concatenate((spcm0, spcm1, spcm0_other, spcm1_other))
    for spcm0, spcm1, spcm0_other, spcm1_other in zip(
        SPCM0_rel_tStamps_iAll,
        SPCM1_rel_tStamps_iAll,
        SPCM0_OtherNode_rel_tStamps_iAll,
        SPCM1_OtherNode_rel_tStamps_iAll,
    )
]

###################################################### plot of photon histograms
if show_photon_hist:
    for i in range(iterations):
        fig, axs = plt.subplots(1, 5, figsize=(18, 3), sharey=True, sharex=True)
    
        axs[0].hist(SPCM0_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[0].set_xlabel("detection time (ns)")
        axs[0].set_ylabel("Occurence")
        axs[0].set_title("SPCM0 photons")
    
        axs[1].hist(SPCM1_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[1].set_xlabel("detection time (ns)")
        axs[1].set_title("SPCM1 photons");
    
        axs[2].hist(SPCM0_OtherNode_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[2].set_xlabel("detection time (ns)")
        axs[2].set_title("SPCM0_OtherNode photons")
    
        axs[3].hist(SPCM1_OtherNode_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7)
        axs[3].set_xlabel("detection time (ns)")
        axs[3].set_title("SPCM1_OtherNode photons");
    
        axs[4].hist(AllSPCMs_rel_tStamps_iAll[i], bins=50, edgecolor='black', alpha=0.7, label=f'{scan_variable1_name}: {scan_sequence1[i]}')
        axs[4].set_xlabel("detection time (ns)")
        axs[4].set_title("All SPCMs photons")
        axs[4].legend(loc=(1.05, 0.2))
##########################################################################################


fig, axs = plt.subplots(1, 3, figsize=(12, 3))
########################################## Plot average number of excitation_cycles per measuremnt
axs[0].plot(scan_sequence1, n_excitation_cycles_iAll/n_measurements, label='ave excitation cycle', marker='o', color='tab:gray')
axs[0].set_xlabel(scan_variable1_name)
axs[0].set_ylabel('n_excitation_cycles')
axs[0].set_title('average n. of excitation cycles')

########################################## Plot g2
axs[1].plot(scan_sequence1, g2_0, marker='o', color='tab:red')
axs[1].errorbar(scan_sequence1, g2_0, yerr=g2_0_err, ls='none',color='tab:red')
axs[1].set_xlabel(scan_variable1_name)
# axs[1].set_ylabel('g2(0)')
axs[1].set_title('g2(0)');

########################################## Plot avergae rate
axs[2].plot(scan_sequence1, Ave_photon_rate_iAll, marker='o', color='tab:red')
# axs[2].legend(loc=(1.05,0.5))
axs[2].set_xlabel(scan_variable1_name)
axs[2].set_ylabel('rate (1/s)')
axs[2].set_title('average rate');

plt.subplots_adjust(wspace=0.4)  # Increase spacing between subplots
plt.show()
```

```python
## coincidence clicks

H1H2_clicks_iAll = np.array([])
V1V2_clicks_iAll = np.array([])
H1V2_clicks_iAll = np.array([])
V1H2_clicks_iAll = np.array([])

H1V1_clicks_iAll = np.array([])
H2V2_clicks_iAll = np.array([])


H1orV1_and_H2orV2_clicks_iAll = np.array([])

H1H2_or_V1V2_clicks_iAll = np.array([])
H1V2_or_V1H2_clicks_iAll = np.array([])

for i in range(iterations):
    start = start_indices[i]
    end = end_indices[i]

    SPCM0_tStamps_i = get_tStamps_i(SPCM0_Photon_tStamps, start, end)                  # H1
    SPCM1_tStamps_i = get_tStamps_i(SPCM1_Photon_tStamps, start, end)                  # V1
    SPCM0_OtherNode_tStamps_i = get_tStamps_i(SPCM0_OtherNode_Photon_tStamps, start, end)  # H2
    SPCM1_OtherNode_tStamps_i = get_tStamps_i(SPCM1_OtherNode_Photon_tStamps, start, end)  # V2
    t1_reference_i = reference_tStamps_t1[1:][start:end]

    ########### checking H1H2 ###############
    count = 0
    for j in range(len(SPCM0_tStamps_i)):
        if SPCM0_tStamps_i[:, 0][j] != -1 and SPCM0_OtherNode_tStamps_i[:, 0][j] != -1:
            count += 1
    H1H2_clicks_iAll = np.append(H1H2_clicks_iAll, count)

    ########### checking V1V2 ###############
    count = 0
    for j in range(len(SPCM1_tStamps_i)):
        if SPCM1_tStamps_i[:, 0][j] != -1 and SPCM1_OtherNode_tStamps_i[:, 0][j] != -1:
            count += 1
    V1V2_clicks_iAll = np.append(V1V2_clicks_iAll, count)

    ########### checking H1V2 ###############
    count = 0
    for j in range(len(SPCM0_tStamps_i)):
        if SPCM0_tStamps_i[:, 0][j] != -1 and SPCM1_OtherNode_tStamps_i[:, 0][j] != -1:
            count += 1
    H1V2_clicks_iAll = np.append(H1V2_clicks_iAll, count)

    ########### checking V1H2 ###############
    count = 0
    for j in range(len(SPCM1_tStamps_i)):
        if SPCM1_tStamps_i[:, 0][j] != -1 and SPCM0_OtherNode_tStamps_i[:, 0][j] != -1:
            count += 1
    V1H2_clicks_iAll = np.append(V1H2_clicks_iAll, count)


    ########### checking H1V1 ###############
    count = 0
    for j in range(len(SPCM0_tStamps_i)):
        if SPCM0_tStamps_i[:, 0][j] != -1 and SPCM1_tStamps_i[:, 0][j] != -1:
            count += 1
    H1V1_clicks_iAll = np.append(H1V1_clicks_iAll, count)

    ########### checking H2V2 ###############
    count = 0
    for j in range(len(SPCM0_tStamps_i)):
        if SPCM0_OtherNode_tStamps_i[:, 0][j] != -1 and SPCM1_OtherNode_tStamps_i[:, 0][j] != -1:
            count += 1
    H2V2_clicks_iAll = np.append(H2V2_clicks_iAll, count)


    ########### checking (H1 or V1) and (H2 or V2) ###############
    count = 0
    for j in range(len(SPCM1_tStamps_i)):
        if (SPCM0_tStamps_i[:, 0][j] != -1 or SPCM1_tStamps_i[:, 0][j] != -1) and \
           (SPCM0_OtherNode_tStamps_i[:, 0][j] != -1 or SPCM1_OtherNode_tStamps_i[:, 0][j] != -1):
            count += 1
    H1orV1_and_H2orV2_clicks_iAll = np.append(H1orV1_and_H2orV2_clicks_iAll, count)

H1H2_or_V1V2_clicks_iAll = H1H2_clicks_iAll + V1V2_clicks_iAll
H1V2_or_V1H2_clicks_iAll = H1V2_clicks_iAll + V1H2_clicks_iAll
H1V1_or_H2V2_clicks_iAll = H1V1_clicks_iAll + H2V2_clicks_iAll

x_scan = np.asarray(scan_sequence1) - 189

fig, ax = plt.subplots(1, 1, figsize=(6, 4))

ax.plot(x_scan, H1H2_clicks_iAll, label='H1H2', marker='o')
ax.plot(x_scan, V1V2_clicks_iAll, label='V1V2', marker='o')
ax.plot(x_scan, H1V2_clicks_iAll, label='H1V2', marker='o')
ax.plot(x_scan, V1H2_clicks_iAll, label='V1H2', marker='o')
# ax.plot(x_scan, H1orV1_and_H2orV2_clicks_iAll, label='(H1 or V1) and (H2 or V2)', marker='o')
# ax.plot(x_scan, H1H2_or_V1V2_clicks_iAll, label='H1H2 + V1V2', marker='o')
# ax.plot(x_scan, H1V2_or_V1H2_clicks_iAll, label='H1V2 + V1H2', marker='o')

ax.plot(x_scan, H1V1_clicks_iAll, label='H1V2', marker='o')
ax.plot(x_scan, H2V2_clicks_iAll, label='V1H2', marker='o')

ax.set_xlabel(f"Timing difference in ns")
ax.set_ylabel('counts')
ax.set_title('Two-node photon coincidence counts')
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()

fig, ax = plt.subplots(1, 1, figsize=(10, 5))

plot_data = [
    (H1orV1_and_H2orV2_clicks_iAll, '(H1 or V1) and (H2 or V2)'),
    (H1H2_or_V1V2_clicks_iAll, 'H1H2 + V1V2'),
    (H1V2_or_V1H2_clicks_iAll, 'H1V2 + V1H2 [Psi-]'),
    (H1V1_or_H2V2_clicks_iAll, 'H1V1 + H2V2 [Psi+]'),
]

for y, label in plot_data:
    line, = ax.plot(x_scan, y, label=label, marker='o')

    ax.errorbar(
        x_scan, y,
        yerr=poisson_err(y),
        fmt='none',
        ecolor=line.get_color(),
        capsize=3,
        label='_nolegend_'
    )

ax.set_xlabel("Timing difference in ns")
ax.set_ylabel('counts')
ax.set_title('Accumulated two-node photon coincidence counts')

ax.legend(loc='center left', bbox_to_anchor=(1.02, 0.5))
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
N_common = np.mean(n_excitation_cycles_iAll)

def scale_to_common_cycles(counts, n_cycles, N_common=None):
    counts = np.asarray(counts, dtype=float)
    n_cycles = np.asarray(n_cycles, dtype=float)

    if N_common is None:
        N_common = np.mean(n_cycles)

    y_scaled = counts / n_cycles * N_common
    yerr_scaled = np.sqrt(counts) / n_cycles * N_common

    return y_scaled, yerr_scaled

fig, ax = plt.subplots(1, 1, figsize=(10, 5))

N_common = np.mean(n_excitation_cycles_iAll)

for counts, label in plot_data:
    y_scaled, yerr_scaled = scale_to_common_cycles(
        counts, n_excitation_cycles_iAll, N_common=N_common
    )

    line, = ax.plot(x_scan, y_scaled, label=label, marker='o')

    ax.errorbar(
        x_scan, y_scaled,
        yerr=yerr_scaled,
        fmt='none',
        ecolor=line.get_color(),
        capsize=3,
        label='_nolegend_'
    )

ax.set_xlabel("Timing difference in ns")
ax.set_ylabel(f"equivalent counts for {N_common:.0f} excitation cycles")
ax.set_title("Two-node photon coincidence counts scaled to common excitation cycles")

ax.legend(loc='center left', bbox_to_anchor=(1.02, 0.5))
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## combining files

### helper functions

```python
def get_full_path_from_index(file_index):
    f = fnames[file_index]
    return f if os.path.isabs(f) else os.path.join(results, f)


def read_h5_datasets(file_index):
    fname = get_full_path_from_index(file_index)

    data = {}
    with h5py.File(fname, "r") as h5:
        ds = h5["datasets"]
        for key in ds.keys():
            x = ds[key][()]
            if isinstance(x, bytes):
                x = x.decode()
            data[key] = x

    return data


def scalar_if_possible(x):
    x = np.asarray(x)
    if x.shape == ():
        return x.item()
    if x.size == 1:
        return x.ravel()[0].item()
    return x


def get_n_excitation_cycles_iAll(n_excitation_cycles, iterations, n_measurements):
    """
    Same logic as the working individual-file analysis:
        n_excitation_cycles_iAll[i] = sum(n_excitation_cycles[i*n_measurements+1:(i+1)*n_measurements+1] + 1)
    """
    n_excitation_cycles = np.asarray(n_excitation_cycles).ravel()
    n_excitation_cycles_iAll = np.zeros(iterations, dtype=int)

    for i in range(iterations):
        n_excitation_cycles_iAll[i] = np.sum(
            np.asarray(
                n_excitation_cycles[i*n_measurements+1:(i+1)*n_measurements+1],
                dtype=int,
            ) + 1
        )

    return n_excitation_cycles_iAll


def get_n_measurements_for_file(data, iterations):
    if "measurements" in data:
        return int(scalar_if_possible(data["measurements"]))

    if "n_measurements" in data:
        return int(scalar_if_possible(data["n_measurements"]))

    if "n_excitation_cycles" in data:
        return len(np.asarray(data["n_excitation_cycles"]).ravel()) // iterations

    raise KeyError("No measurements / n_measurements dataset found in this file.")


def get_n_excitation_cycles_iAll_from_data(data, iterations=None):
    """
    Return excitation cycles per scan point for THIS file.
    This avoids accidentally using a global n_excitation_cycles_iAll from another file.
    """
    if iterations is None:
        scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
        iterations = len(scan_sequence1)

    if "n_excitation_cycles_iAll" in data:
        return np.asarray(data["n_excitation_cycles_iAll"], dtype=int).ravel()

    if "n_excitation_cycles" in data:
        n_excitation_cycles = np.asarray(data["n_excitation_cycles"]).ravel()
        n_measurements_file = get_n_measurements_for_file(data, iterations)

        if n_excitation_cycles.size == 1:
            # Fallback for files that store only one constant value.
            return np.full(
                iterations,
                int(scalar_if_possible(n_excitation_cycles)) * n_measurements_file,
                dtype=int,
            )

        return get_n_excitation_cycles_iAll(
            n_excitation_cycles,
            iterations,
            n_measurements_file,
        )

    raise KeyError("No excitation-cycle dataset found in this file.")


def get_start_end_indices_for_file(data, n_excitation_cycles_iAll_file=None):
    """
    IMPORTANT FIX.

    Do NOT split the timestamp arrays into equal-sized chunks.
    Individual-file analysis uses:
        n_tStamps_iAll = n_excitation_cycles_iAll * n_excitation_attempts
        end_indices = np.cumsum(n_tStamps_iAll)
        start_indices = np.insert(end_indices[:-1], 0, 0)

    The accumulated-file analysis has to use the same boundaries.
    Otherwise photon timestamps from one scan point can be subtracted from
    reference timestamps belonging to a neighboring scan point, which creates
    fake out-of-window arrival times.
    """
    scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
    iterations = len(scan_sequence1)

    if n_excitation_cycles_iAll_file is None:
        n_excitation_cycles_iAll_file = get_n_excitation_cycles_iAll_from_data(
            data,
            iterations,
        )

    if "n_excitation_attempts" in data:
        n_excitation_attempts_file = int(scalar_if_possible(data["n_excitation_attempts"]))
    else:
        n_excitation_attempts_file = 1

    n_tStamps_iAll_file = np.asarray(n_excitation_cycles_iAll_file, dtype=int) * n_excitation_attempts_file

    end_indices = np.cumsum(n_tStamps_iAll_file)
    start_indices = np.insert(end_indices[:-1], 0, 0)

    n_total_reference = len(data["reference_tStamps_t1"]) - 1
    if int(end_indices[-1]) != int(n_total_reference):
        print(
            "WARNING: timestamp-boundary mismatch for this file. "
            f"sum(n_tStamps_iAll) = {int(end_indices[-1])}, "
            f"len(reference_tStamps_t1)-1 = {int(n_total_reference)}. "
            "Check n_excitation_cycles_iAll / n_excitation_cycles / n_excitation_attempts."
        )

    return start_indices, end_indices


def rel_timestamps_ns(tStamps_i, t1_reference_i):
    """
    Convert timestamp array for one scan point into relative detection times in ns.

    IMPORTANT:
    This now matches the individual-file analysis logic in get_rel_tStamps_i():
      - remove -1 no-click entries
      - subtract the matching t1 reference for that same shot
      - flatten 1-click and 2-click timestamps
      - remove negative relative timestamps

    The previous accumulated-file helper kept negative relative timestamps.
    That made the accumulated histogram look different from the individual-file histogram.
    """
    tStamps_i = np.asarray(tStamps_i)
    t1_reference_i = np.asarray(t1_reference_i)

    if tStamps_i.size == 0 or t1_reference_i.size == 0:
        return np.array([], dtype=float)

    # Make the shape shot x click_index.
    # For the normal max_clicks=2 case this is already 2D.
    if tStamps_i.ndim == 1:
        tStamps_i = tStamps_i.reshape(-1, 1)

    # Match the original individual-file logic, which uses zip(...)
    # and therefore only compares rows with existing references.
    n = min(len(tStamps_i), len(t1_reference_i))
    tStamps_i = tStamps_i[:n]
    t1_reference_i = t1_reference_i[:n]

    rel_chunks = [
        row[row != -1] - ref
        for row, ref in zip(tStamps_i, t1_reference_i)
    ]

    if len(rel_chunks) == 0:
        return np.array([], dtype=float)

    rel_tStamps_i = np.concatenate(rel_chunks) * 1e9
    rel_tStamps_i = rel_tStamps_i[np.isfinite(rel_tStamps_i)]

    # Same filter as the individual-file analysis:
    # in rare cases a photon timestamp is earlier than its reference.
    # Those were removed in the working individual-file path.
    rel_tStamps_i = rel_tStamps_i[rel_tStamps_i >= 0]

    return rel_tStamps_i


def first_click_exists(tStamps_i):
    tStamps_i = np.asarray(tStamps_i)

    if tStamps_i.ndim == 1:
        return tStamps_i != -1

    return tStamps_i[:, 0] != -1


def analyze_one_file(file_index):
    data = read_h5_datasets(file_index)

    scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
    iterations = len(scan_sequence1)

    # Get excitation cycles first, then use them to compute the SAME
    # start/end indices as the working individual-file analysis.
    n_excitation_cycles_iAll_file = get_n_excitation_cycles_iAll_from_data(data, iterations)
    start_indices, end_indices = get_start_end_indices_for_file(data, n_excitation_cycles_iAll_file)

    if "Atom_loading_time" in data:
        all_atom_loading_times = [np.asarray(data["Atom_loading_time"]).ravel()[1:]]
    else:
        all_atom_loading_times = []

    SPCM0_rel_tStamps_iAll = []
    SPCM1_rel_tStamps_iAll = []
    SPCM0_OtherNode_rel_tStamps_iAll = []
    SPCM1_OtherNode_rel_tStamps_iAll = []
    AllSPCMs_rel_tStamps_iAll = []

    H1H2_clicks_iAll = np.array([])
    V1V2_clicks_iAll = np.array([])
    H1V2_clicks_iAll = np.array([])
    V1H2_clicks_iAll = np.array([])

    H1V1_clicks_iAll = np.array([])
    H2V2_clicks_iAll = np.array([])

    H1orV1_and_H2orV2_clicks_iAll = np.array([])

    for i in range(iterations):
        start = start_indices[i]
        end = end_indices[i]

        SPCM0_tStamps_i = get_tStamps_i(
            data["SPCM0_Photon_tStamps"],
            start,
            end,
        )  # H1

        SPCM1_tStamps_i = get_tStamps_i(
            data["SPCM1_Photon_tStamps"],
            start,
            end,
        )  # V1

        SPCM0_OtherNode_tStamps_i = get_tStamps_i(
            data["SPCM0_OtherNode_Photon_tStamps"],
            start,
            end,
        )  # H2

        SPCM1_OtherNode_tStamps_i = get_tStamps_i(
            data["SPCM1_OtherNode_Photon_tStamps"],
            start,
            end,
        )  # V2

        t1_reference_i = data["reference_tStamps_t1"][1:][start:end]

        # Compute the relative timestamp arrays detector-by-detector.
        # Then build AllSPCMs_rel_tStamps_iAll from these four arrays.
        SPCM0_rel_i = rel_timestamps_ns(SPCM0_tStamps_i, t1_reference_i)
        SPCM1_rel_i = rel_timestamps_ns(SPCM1_tStamps_i, t1_reference_i)
        SPCM0_OtherNode_rel_i = rel_timestamps_ns(SPCM0_OtherNode_tStamps_i, t1_reference_i)
        SPCM1_OtherNode_rel_i = rel_timestamps_ns(SPCM1_OtherNode_tStamps_i, t1_reference_i)

        SPCM0_rel_tStamps_iAll.append(SPCM0_rel_i)
        SPCM1_rel_tStamps_iAll.append(SPCM1_rel_i)
        SPCM0_OtherNode_rel_tStamps_iAll.append(SPCM0_OtherNode_rel_i)
        SPCM1_OtherNode_rel_tStamps_iAll.append(SPCM1_OtherNode_rel_i)

        AllSPCMs_rel_tStamps_iAll.append(
            np.concatenate((
                SPCM0_rel_i,
                SPCM1_rel_i,
                SPCM0_OtherNode_rel_i,
                SPCM1_OtherNode_rel_i,
            ))
        )

        H1 = first_click_exists(SPCM0_tStamps_i)
        V1 = first_click_exists(SPCM1_tStamps_i)
        H2 = first_click_exists(SPCM0_OtherNode_tStamps_i)
        V2 = first_click_exists(SPCM1_OtherNode_tStamps_i)

        n = min(len(H1), len(V1), len(H2), len(V2))
        H1, V1, H2, V2 = H1[:n], V1[:n], H2[:n], V2[:n]

        H1H2_clicks_iAll = np.append(H1H2_clicks_iAll, np.sum(H1 & H2))
        V1V2_clicks_iAll = np.append(V1V2_clicks_iAll, np.sum(V1 & V2))
        H1V2_clicks_iAll = np.append(H1V2_clicks_iAll, np.sum(H1 & V2))
        V1H2_clicks_iAll = np.append(V1H2_clicks_iAll, np.sum(V1 & H2))

        H1V1_clicks_iAll = np.append(H1V1_clicks_iAll, np.sum(H1 & V1))
        H2V2_clicks_iAll = np.append(H2V2_clicks_iAll, np.sum(H2 & V2))

        H1orV1_and_H2orV2_clicks_iAll = np.append(
            H1orV1_and_H2orV2_clicks_iAll,
            np.sum((H1 | V1) & (H2 | V2)),
        )

    return {
        "file_index": file_index,
        "scan_sequence1": scan_sequence1,
        "scan_variable1_name": data["scan_variable1_name"],

        "n_excitation_cycles_iAll": n_excitation_cycles_iAll_file,

        "SPCM0_rel_tStamps_iAll": SPCM0_rel_tStamps_iAll,
        "SPCM1_rel_tStamps_iAll": SPCM1_rel_tStamps_iAll,
        "SPCM0_OtherNode_rel_tStamps_iAll": SPCM0_OtherNode_rel_tStamps_iAll,
        "SPCM1_OtherNode_rel_tStamps_iAll": SPCM1_OtherNode_rel_tStamps_iAll,
        "AllSPCMs_rel_tStamps_iAll": AllSPCMs_rel_tStamps_iAll,

        "H1H2": H1H2_clicks_iAll,
        "V1V2": V1V2_clicks_iAll,
        "H1V2": H1V2_clicks_iAll,
        "V1H2": V1H2_clicks_iAll,

        "H1V1": H1V1_clicks_iAll,
        "H2V2": H2V2_clicks_iAll,

        "H1orV1_and_H2orV2": H1orV1_and_H2orV2_clicks_iAll,

        "all_atom_loading_times": all_atom_loading_times,
    }


import re
from IPython.display import display, HTML


def get_rid_from_file(file_index):
    fname = get_full_path_from_index(file_index)

    rid = None

    with h5py.File(fname, "r") as h5:
        # Try h5 attributes first
        for key in ["rid", "RID", "run_id"]:
            if key in h5.attrs:
                rid = h5.attrs[key]
                break

        # Try datasets group
        if rid is None and "datasets" in h5:
            ds = h5["datasets"]
            for key in ["rid", "RID", "run_id"]:
                if key in ds:
                    rid = ds[key][()]
                    break

    if isinstance(rid, bytes):
        rid = rid.decode()

    if rid is not None:
        return scalar_if_possible(rid)

    # Fallback: try to extract RID from filename
    base = os.path.basename(fname)
    match = re.search(r"(?:RID|rid)[_\-\s]*(\d+)", base)

    if match is None:
        match = re.search(r"(\d{4,})", base)

    return match.group(1) if match else "unknown"


def get_file_summary_row(file_index):
    data = read_h5_datasets(file_index)

    scan_sequence1 = np.asarray(data["scan_sequence1"]).ravel()
    iterations = len(scan_sequence1)

    if "measurements" in data:
        measurements = scalar_if_possible(data["measurements"])
    elif "n_measurements" in data:
        measurements = scalar_if_possible(data["n_measurements"])
    else:
        measurements = "unknown"

    try:
        n_excitation_cycles_iAll_file = get_n_excitation_cycles_iAll_from_data(data, iterations)
        total_excitation_cycles = int(np.sum(n_excitation_cycles_iAll_file))
    except Exception:
        total_excitation_cycles = len(data["reference_tStamps_t1"]) - 1

    return {
        "file_index": file_index,
        "RID": get_rid_from_file(file_index),
        "# measurements": measurements,
        "# total excitation cycles": total_excitation_cycles,
    }


def display_selected_file_summary(selected_file_indices):
    rows = [get_file_summary_row(file_index) for file_index in selected_file_indices]

    numeric_measurements = [row["# measurements"] for row in rows if isinstance(row["# measurements"], (int, float, np.integer, np.floating))]
    total_measurements = np.sum(numeric_measurements) if len(numeric_measurements) else "unknown"
    total_excitation_cycles = np.sum([row["# total excitation cycles"] for row in rows])

    html = """
    <table>
        <tr>
            <th>file_index</th>
            <th>RID</th>
            <th># measurements</th>
            <th># total excitation cycles</th>
        </tr>
    """

    for row in rows:
        html += f"""
        <tr>
            <td>{row["file_index"]}</td>
            <td>{row["RID"]}</td>
            <td>{row["# measurements"]}</td>
            <td>{row["# total excitation cycles"]}</td>
        </tr>
        """

    html += f"""
        <tr>
            <td><b>TOTAL</b></td>
            <td></td>
            <td><b>{total_measurements}</b></td>
            <td><b>{total_excitation_cycles}</b></td>
        </tr>
    """

    html += "</table>"

    display(HTML(html))

    return rows
```

### select files and combine

```python
# ============================================================
# Select files to accumulate
# ============================================================

# #### HOM -100, 0, 100
selected_file_indices = [
    # 4, 5, 6, 9, 10, 11,
    # 14, 15, 17, 19, 21, 
    # 23, 25, 12,
    27, 29,
    # 23, 25
]

selected_file_summary = display_selected_file_summary(selected_file_indices)

# ============================================================
# Analyze selected files
# ============================================================

file_results = [analyze_one_file(file_index) for file_index in selected_file_indices]

scan_sequence1 = file_results[0]["scan_sequence1"]
scan_variable1_name = file_results[0]["scan_variable1_name"]
iterations = len(scan_sequence1)

# Check that all selected files have the same scan sequence
for r in file_results[1:]:
    if len(r["scan_sequence1"]) != len(scan_sequence1) or not np.allclose(r["scan_sequence1"], scan_sequence1):
        raise ValueError(
            f"File index {r['file_index']} does not have the same scan_sequence1."
        )

# ============================================================
# Accumulate detector-specific relative timestamps first
# ============================================================

SPCM0_rel_tStamps_iAll = [
    np.concatenate([r["SPCM0_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

SPCM1_rel_tStamps_iAll = [
    np.concatenate([r["SPCM1_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

SPCM0_OtherNode_rel_tStamps_iAll = [
    np.concatenate([r["SPCM0_OtherNode_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

SPCM1_OtherNode_rel_tStamps_iAll = [
    np.concatenate([r["SPCM1_OtherNode_rel_tStamps_iAll"][i] for r in file_results])
    for i in range(iterations)
]

# Now build AllSPCMs_rel_tStamps_iAll in exactly the same way as the
# individual-file analysis.
AllSPCMs_rel_tStamps_iAll = [
    np.concatenate((spcm0, spcm1, spcm0_other, spcm1_other))
    for spcm0, spcm1, spcm0_other, spcm1_other in zip(
        SPCM0_rel_tStamps_iAll,
        SPCM1_rel_tStamps_iAll,
        SPCM0_OtherNode_rel_tStamps_iAll,
        SPCM1_OtherNode_rel_tStamps_iAll,
    )
]

# Diagnostic: this should now reflect only true relative timestamps.
for i, arr in enumerate(AllSPCMs_rel_tStamps_iAll):
    if len(arr) > 0:
        print(
            f"scan i={i}, scan value={scan_sequence1[i]}: "
            f"N={len(arr)}, min={np.min(arr):.3f} ns, max={np.max(arr):.3f} ns"
        )
    else:
        print(f"scan i={i}, scan value={scan_sequence1[i]}: N=0")

# ============================================================
# Accumulate coincidence counts
# ============================================================

H1H2_clicks_iAll = np.sum([r["H1H2"] for r in file_results], axis=0)
V1V2_clicks_iAll = np.sum([r["V1V2"] for r in file_results], axis=0)
H1V2_clicks_iAll = np.sum([r["H1V2"] for r in file_results], axis=0)
V1H2_clicks_iAll = np.sum([r["V1H2"] for r in file_results], axis=0)

H1V1_clicks_iAll = np.sum([r["H1V1"] for r in file_results], axis=0)
H2V2_clicks_iAll = np.sum([r["H2V2"] for r in file_results], axis=0)

H1orV1_and_H2orV2_clicks_iAll = np.sum(
    [r["H1orV1_and_H2orV2"] for r in file_results],
    axis=0,
)

H1H2_or_V1V2_clicks_iAll = H1H2_clicks_iAll + V1V2_clicks_iAll
H1V2_or_V1H2_clicks_iAll = H1V2_clicks_iAll + V1H2_clicks_iAll
H1V1_or_H2V2_clicks_iAll = H1V1_clicks_iAll + H2V2_clicks_iAll

# ============================================================
# Accumulate excitation cycles
# ============================================================

n_excitation_cycles_iAll_combined = np.sum(
    [r["n_excitation_cycles_iAll"] for r in file_results],
    axis=0,
)

# Keep this name also, because later plotting cells may expect it.
n_excitation_cycles_iAll = n_excitation_cycles_iAll_combined

# ============================================================
# Accumulate atom loading times, if present
# ============================================================

all_atom_loading_times = []
for r in file_results:
    all_atom_loading_times.extend(r.get("all_atom_loading_times", []))

print("Accumulated file indices:", selected_file_indices)
print("scan_sequence1:", scan_sequence1)
print("n_excitation_cycles_iAll_combined:", n_excitation_cycles_iAll_combined)
```

```python
###################################################### plot of accumulated photon histograms

ncols = 7
nrows = int(np.ceil(iterations / ncols))

fig, axs = plt.subplots(nrows, ncols, figsize=(18, 3*nrows), sharey=True, sharex=True)
axs = np.ravel(axs)

for i in range(iterations):
    axs[i].hist(
        AllSPCMs_rel_tStamps_iAll[i],
        bins=30,
        edgecolor='black',
        alpha=0.7,
        label=f'{scan_variable1_name}: {scan_sequence1[i]}'
    )

    axs[i].set_xlabel("detection time (ns)")
    axs[i].set_title(f"i = {i}")
    axs[i].legend(fontsize=8)
    axs[i].tick_params(axis='x', labelbottom=True)
    # axs[i].set_xlim(600, 1000)

for j in range(iterations, len(axs)):
    axs[j].axis("off")

fig.suptitle("Accumulated All SPCMs photons", fontsize=14)
plt.tight_layout()
plt.show()
```

```python

# ============================================================
# Diagnostic: show remaining out-of-window timestamps after accumulation
# ============================================================
# This does NOT filter anything. It only reports what remains.

try:
    collection_window_ns = float(t_photon_collection_time) * 1e9
except Exception:
    collection_window_ns = None

print("Accumulated relative-timestamp diagnostic")
if collection_window_ns is not None:
    print(f"collection_window_ns = {collection_window_ns:.3f} ns")
else:
    print("collection_window_ns = unknown because t_photon_collection_time was not found")

for i, arr in enumerate(AllSPCMs_rel_tStamps_iAll):
    arr = np.asarray(arr, dtype=float)
    if arr.size == 0:
        print(f"i={i}, scan={scan_sequence1[i]}: N=0")
        continue

    n_neg = np.sum(arr < 0)
    if collection_window_ns is None:
        n_late = 0
    else:
        n_late = np.sum(arr > collection_window_ns)

    print(
        f"i={i}, scan={scan_sequence1[i]}: "
        f"N={arr.size}, min={np.min(arr):.3f} ns, max={np.max(arr):.3f} ns, "
        f"negative={n_neg}, late={n_late}"
    )
```

```python
# ============================================================
# Plot equivalent counts scaled to common excitation cycles
# ============================================================

def scale_to_common_cycles(counts, n_cycles, N_common=None):
    counts = np.asarray(counts, dtype=float)
    n_cycles = np.asarray(n_cycles, dtype=float)

    if N_common is None:
        N_common = np.mean(n_cycles)

    y_scaled = counts / n_cycles * N_common
    yerr_scaled = np.sqrt(counts) / n_cycles * N_common

    return y_scaled, yerr_scaled


plot_data = [
    (H1orV1_and_H2orV2_clicks_iAll, '(H1 or V1) and (H2 or V2)'),
    (H1H2_or_V1V2_clicks_iAll, 'H1H2 + V1V2'),
    (H1V2_or_V1H2_clicks_iAll, 'H1V2 + V1H2 [Psi-]'),
    (H1V1_or_H2V2_clicks_iAll, 'H1V1 + H2V2 [Psi+]'),
]

x_scan = scan_sequence1 - 189  # keep this if this is your timing-zero convention

N_common = np.mean(n_excitation_cycles_iAll_combined)

fig, ax = plt.subplots(1, 1, figsize=(10, 5))

for counts, label in plot_data:
    y_scaled, yerr_scaled = scale_to_common_cycles(
        counts,
        n_excitation_cycles_iAll_combined,
        N_common=N_common
    )

    line, = ax.plot(x_scan, y_scaled, label=label, marker='o')

    ax.errorbar(
        x_scan,
        y_scaled,
        yerr=yerr_scaled,
        fmt='none',
        ecolor=line.get_color(),
        capsize=3,
        label='_nolegend_'
    )

ax.set_xlabel("Timing difference in ns")
ax.set_ylabel(f"equivalent counts for {N_common:.0f} excitation cycles")
ax.set_title("Two-node photon coincidence counts scaled to common excitation cycles")

ax.legend(loc='center left', bbox_to_anchor=(1.02, 0.5))
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
